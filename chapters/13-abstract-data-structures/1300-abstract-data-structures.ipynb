# Abstract Data Structures

Abstract data structures describe how collections behave before we decide exactly how to implement them.

- Abstract data types define supported operations and expected behavior
- Concrete data structures store values in specific ways
- Operation costs guide structure choice
- Stacks and queues model order-sensitive workflows
- Trees and graphs model hierarchical and relational data

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Abstract data type | A behavior-focused description of a collection and its operations. |
| 2 | Concrete implementation | The storage and code choices used to perform an abstract data type's behavior. |
| 3 | Operation | An action a data structure supports, such as access, search, insert, or delete. |
| 4 | Access | Retrieving an item from a data structure, often by position or key. |
| 5 | Search | Looking through a data structure to find a value or relationship. |
| 6 | Insert | Adding a new item to a data structure. |
| 7 | Delete | Removing an item from a data structure. |
| 8 | Tradeoff | A design choice where improving one operation or representation can make another cost worse. |
| 9 | Stack | A last-in, first-out collection. |
| 10 | LIFO | Last in, first out, the access rule used by stacks. |
| 11 | Push | The stack operation that adds an item to the top. |
| 12 | Pop | The stack operation that removes and returns the top item. |
| 13 | Peek | The stack operation that looks at the top item without removing it. |
| 14 | Queue | A first-in, first-out collection. |
| 15 | FIFO | First in, first out, the access rule used by queues. |
| 16 | Enqueue | The queue operation that adds an item to the back. |
| 17 | Dequeue | The queue operation that removes and returns the front item. |
| 18 | Tree | A hierarchical structure with a root and child nodes. |
| 19 | Graph | A structure made of nodes and edges representing relationships. |
| 20 | Traversal | A systematic process for visiting nodes in a data structure. |
