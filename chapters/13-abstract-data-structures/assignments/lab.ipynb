```python
### Question 1: Stack Class: Browsing History
# Implement a Stack class backed by a list, with:
# - append(item): push item onto the top of the stack
# - pop(): remove and return the top item (LIFO order)
# - __len__: number of items currently on the stack
# Push 'home', then 'products', then 'checkout' onto a Stack named history.
# Pop twice and print each popped value, then print len(history).
# Print exactly these three lines:
# checkout
# products
# 1
### Your code starts here.



### Your code ends here.
```

```python
### Solution
class Stack:
    def __init__(self):
        self.items = []

    def append(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def __len__(self):
        return len(self.items)

history = Stack()
history.append("home")
history.append("products")
history.append("checkout")

print(history.pop())
print(history.pop())
print(len(history))
```

```python
### Question 2: Queue Class: Serve a Customer Line
# Implement a Queue class backed by a list, with:
# - append(item): add item to the back of the line
# - pop(): remove and return the item at the front (FIFO order) --
#   use self.items.pop(0) so removal happens from the front
# - __len__: number of items currently waiting
# Add 'apple', 'banana', 'cherry' to a Queue named customers.
# Serve two customers by calling pop() twice and print each result,
# then print len(customers). Print exactly these three lines:
# apple
# banana
# 1
### Your code starts here.



### Your code ends here.
```

```python
### Solution
class Queue:
    def __init__(self):
        self.items = []

    def append(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop(0)

    def __len__(self):
        return len(self.items)

customers = Queue()
customers.append("apple")
customers.append("banana")
customers.append("cherry")

print(customers.pop())
print(customers.pop())
print(len(customers))
```

```python
### Question 3: Add a Node to an Org-Chart Tree
# org_chart is a nested dict tree: each node has 'name' and 'children'.
# Write add_employee(node, manager_name, new_name) that searches the
# tree recursively for the node whose name is manager_name, and, once
# found, appends a new node {'name': new_name, 'children': []} to
# that manager's children list. Return True once added, False if the
# manager was never found (search the rest of the tree recursively
# otherwise).
# Write list_names(node) that returns a flat list of every name in
# the tree, visiting a node before its children (root first).
# Call add_employee(org_chart, 'charlie', 'erin'), then print
# list_names(org_chart) exactly:
# ['operations', 'charlie', 'erin', 'dana']
org_chart = {
    "name": "operations",
    "children": [
        {"name": "charlie", "children": []},
        {"name": "dana", "children": []},
    ],
}
### Your code starts here.



### Your code ends here.
```

```python
### Solution
org_chart = {
    "name": "operations",
    "children": [
        {"name": "charlie", "children": []},
        {"name": "dana", "children": []},
    ],
}

def add_employee(node, manager_name, new_name):
    if node["name"] == manager_name:
        node["children"].append({"name": new_name, "children": []})
        return True
    for child in node["children"]:
        if add_employee(child, manager_name, new_name):
            return True
    return False

def list_names(node):
    names = [node["name"]]
    for child in node["children"]:
        names.extend(list_names(child))
    return names

add_employee(org_chart, "charlie", "erin")
print(list_names(org_chart))
```

```python
### Question 4: Breadth-First Traversal of a Graph
# graph is an adjacency-list dict mapping each person to their
# connections. Write bfs_order(graph, start) that performs a
# breadth-first traversal starting at start and returns the list of
# names in the order they were first visited.
# Use a plain list as the queue (append to enqueue, pop(0) to
# dequeue) and a plain list named visited to track visited names
# (use visited.count(name) == 0 to test whether a name is new).
# Call bfs_order(graph, 'alice') and print exactly:
# ['alice', 'bob', 'charlie', 'dana', 'erin']
graph = {
    "alice": ["bob", "charlie"],
    "bob": ["alice", "dana"],
    "charlie": ["alice", "dana"],
    "dana": ["bob", "charlie", "erin"],
    "erin": ["dana"],
}
### Your code starts here.



### Your code ends here.
```

```python
### Solution
graph = {
    "alice": ["bob", "charlie"],
    "bob": ["alice", "dana"],
    "charlie": ["alice", "dana"],
    "dana": ["bob", "charlie", "erin"],
    "erin": ["dana"],
}

def bfs_order(graph, start):
    visited = []
    order = []
    queue = [start]
    while queue:
        node = queue.pop(0)
        if visited.count(node) == 0:
            visited.append(node)
            order.append(node)
            for neighbor in graph.get(node, []):
                if visited.count(neighbor) == 0:
                    queue.append(neighbor)
    return order

print(bfs_order(graph, "alice"))
```

<script>
(function () {
  var labId = 'ch13-lab';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';
    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var codeMirror = cell.querySelector('.CodeMirror');
    if (codeMirror && codeMirror.CodeMirror && typeof codeMirror.CodeMirror.getValue === 'function') {
      return codeMirror.CodeMirror.getValue();
    }
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent) {
      return Array.from(cmContent.querySelectorAll('.cm-line')).map(function (line) { return line.textContent; }).join('\n');
    }
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.from(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 1)] = codeFromCell(cell);
    });
    return code;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch13-lab-feedback-list');
    if (list) list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row ? row.querySelector('span') : null;
      var item = feedback ? feedback[question] : null;
      if (!row || !message || !item) return;
      row.classList.remove('correct', 'incorrect');
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch13-lab-feedback-list');
    if (list) list.hidden = true;
    grader.querySelectorAll('.ch13-lab-feedback-row').forEach(function (row) {
      row.classList.remove('correct', 'incorrect');
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var saved = text ? JSON.parse(text) : null;
        if (!response.ok || !saved || !saved.ok) throw new Error(saved && (saved.message || saved.error) ? (saved.message || saved.error) : 'Submission API is unavailable.');
        return saved;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch13-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var result = document.getElementById('ch13-lab-result');
    var submit = document.getElementById('ch13-lab-submit');
    var actions = grader.querySelector('.ch13-lab-actions');
    var reset = document.getElementById('ch13-lab-reset');

    loadSessionIdentity(identity, actions, submit, result);

    submit.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = 'Submitting code.';
      submit.disabled = true;
      saveAttempt({
        lab_id: labId,
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          if (payload.feedback) applyFeedback(grader, payload.feedback);
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
</script>
