# Strings and Text

Strings are sequences of characters, so many sequence ideas from lists and tuples also apply to text.

- Strings as immutable sequences
- String methods for cleaning and transforming text
- Regular expressions for pattern matching
- Dictionaries and counters for text analysis
- Random text and Markov-style generation

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | String | An immutable sequence of text characters. |
| 2 | Escape sequence | A backslash-based notation for characters that are hard to type directly. |
| 3 | Raw string | A string literal that treats backslashes mostly as ordinary characters. |
| 4 | Indexing | Accessing one character from a string by position. |
| 5 | Slicing | Creating a substring from a range of positions. |
| 6 | String method | A function called on a string object. |
| 7 | Case method | A string method that changes or checks capitalization. |
| 8 | Splitting | Breaking one string into a list of smaller strings. |
| 9 | Joining | Combining strings from an iterable into one string. |
| 10 | String formatting | Building strings that include values from variables or expressions. |
| 11 | Docstring | A string at the start of a module, function, class, or method that documents it. |
| 12 | Regular expression | A pattern language for matching text. |
| 13 | `re` module | Python's standard-library module for regular expressions. |
| 14 | Match object | An object that stores information about a successful regular expression match. |
| 15 | Metacharacter | A character with special meaning in a regular expression. |
| 16 | Quantifier | A regular expression symbol that controls how many times a pattern may repeat. |
| 17 | Character class | A regular expression pattern that matches one character from a set of choices. |
| 18 | Counter | A dictionary-like collection for counting hashable values. |
| 19 | Bigram | A sequence of two adjacent words or tokens. |
| 20 | Markov analysis | A text-generation approach that predicts the next item from recent context. |
