- Sequence, decision, and repetition as control patterns
- Runtime measurement and Big O growth

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Algorithm | A step-by-step procedure for solving a problem or completing a task. |
| 2 | Input | The data an algorithm receives before it runs. |
| 3 | Output | The result an algorithm produces. |
| 4 | Correctness | Whether an algorithm produces the right output for valid inputs. |
| 5 | Complexity | A description of how an algorithm's resource use grows with input size. |
| 6 | Time complexity | How running time grows as input size grows. |
| 7 | Space complexity | How memory use grows as input size grows. |
| 8 | Big O notation | A notation for describing an algorithm's growth rate at a high level. |
| 9 | Constant time | A growth pattern where work does not depend on input size. |
| 10 | Linear time | A growth pattern where work grows in direct proportion to input size. |
| 11 | Logarithmic time | A growth pattern where each step greatly reduces the remaining search space. |
| 12 | Benchmarking | Measuring code performance by running timed experiments. |
| 13 | Linear search | A search algorithm that checks items one at a time. |
| 14 | Binary search | A search algorithm that repeatedly halves a sorted search space. |
| 15 | `bisect` | A standard-library module for searching and inserting in sorted sequences. |
| 16 | Hash-based lookup | Finding values through a hash table, commonly using a `set` or `dict`. |
| 17 | Sorting | Rearranging values into an order. |
| 18 | Stability | A sorting property that preserves the relative order of equal-key items. |
| 19 | In-place sort | A sort that rearranges the original list instead of returning a full separate copy. |
| 20 | Divide and conquer | A design strategy that solves a problem by splitting it into smaller subproblems. |
