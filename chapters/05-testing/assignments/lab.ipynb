```python
### Question 1: Safe conversion with try/except
# Write safe_float(s) that:
# 1. Tries to convert the string s to a float with float(s).
# 2. Returns the float if conversion succeeds.
# 3. If s cannot be converted, catches the exception, prints a descriptive
#    message using repr(s), and returns None.
# Call safe_float("3.14") and safe_float("abc"), printing each result.
# Print exactly these three lines:
# 3.14
# Cannot convert 'abc' to float
# None
### Your code starts here.



### Your code ends here.
```

```python
### Solution
def safe_float(s):
    try:
        return float(s)
    except ValueError:
        print(f"Cannot convert {s!r} to float")
        return None

print(safe_float("3.14"))
print(safe_float("abc"))
```

```python
### Question 2: Assert-based validator
# Add assert statements to validate_rectangle(width, height) that check:
# 1. width and height are both positive (use a single assert for both).
# 2. width and height are both at most 100 (use a single assert for both).
# Each assert should include a descriptive message.
# Call validate_rectangle(10, 20), then catch failures from
# validate_rectangle(-5, 20) and validate_rectangle(10, 500) with
# except Exception as e, printing f"AssertionError: {e}" for each.
# Print exactly these three lines:
# (10, 20) is a valid rectangle
# AssertionError: Width and height must be positive
# AssertionError: Width and height must be at most 100
### Your code starts here.



### Your code ends here.
```

```python
### Solution
def validate_rectangle(width, height):
    assert width > 0 and height > 0, "Width and height must be positive"
    assert width <= 100 and height <= 100, "Width and height must be at most 100"

validate_rectangle(10, 20)
print("(10, 20) is a valid rectangle")

try:
    validate_rectangle(-5, 20)
except Exception as e:
    print(f"AssertionError: {e}")

try:
    validate_rectangle(10, 500)
except Exception as e:
    print(f"AssertionError: {e}")
```

```python
### Question 3: Multiple exception types
# Write safe_percentage(score, total) that:
# 1. Returns score / total * 100.
# 2. Catches ZeroDivisionError (total is 0) -- prints a message, returns None.
# 3. Catches TypeError (score or total is not a number) -- prints a message,
#    returns None.
# Print exactly these five lines:
# 75.0
# Error: total cannot be zero
# None
# Error: score and total must be numbers
# None
### Your code starts here.



### Your code ends here.
```

```python
### Solution
def safe_percentage(score, total):
    try:
        return score / total * 100
    except ZeroDivisionError:
        print("Error: total cannot be zero")
        return None
    except TypeError:
        print("Error: score and total must be numbers")
        return None

print(safe_percentage(45, 60))
print(safe_percentage(45, 0))
print(safe_percentage(45, "sixty"))
```

```python
### Question 4: try/except/else/finally together
# Write safe_lookup(record, field) that:
# 1. Tries value = record[field].
# 2. Catches KeyError -- prints a message naming the missing field, sets
#    value to None.
# 3. Uses else to print "Field found." only when the lookup succeeds.
# 4. Uses finally to always print "Lookup attempt finished."
# 5. Returns value.
# Call safe_lookup(student_record, "grade") and
# safe_lookup(student_record, "email"), printing each result.
# Print exactly these six lines:
# Field found.
# Lookup attempt finished.
# A
# Field 'email' not found
# Lookup attempt finished.
# None
student_record = {"name": "Priya", "grade": "A"}
### Your code starts here.



### Your code ends here.
```

```python
### Solution
def safe_lookup(record, field):
    try:
        value = record[field]
    except KeyError:
        print(f"Field {field!r} not found")
        value = None
    else:
        print("Field found.")
    finally:
        print("Lookup attempt finished.")
    return value

student_record = {"name": "Priya", "grade": "A"}

print(safe_lookup(student_record, "grade"))
print(safe_lookup(student_record, "email"))
```

```python
### Question 5: Raising exceptions by condition
# Write check_password_strength(password) that:
# 1. Raises TypeError with a descriptive message (including type(password))
#    if password is not a string.
# 2. Raises ValueError with a descriptive message if password has fewer
#    than 8 characters.
# 3. Otherwise returns True.
# Call check_password_strength("longenough") and print the result.
# Then catch failures from check_password_strength("short") and
# check_password_strength(12345678), printing each caught message.
# Print exactly these three lines:
# True
# ValueError: Password must be at least 8 characters
# TypeError: Password must be a string, got <class 'int'>
### Your code starts here.



### Your code ends here.
```

```python
### Solution
def check_password_strength(password):
    if not isinstance(password, str):
        raise TypeError(f"Password must be a string, got {type(password)}")
    if len(password) < 8:
        raise ValueError("Password must be at least 8 characters")
    return True

print(check_password_strength("longenough"))

try:
    check_password_strength("short")
except ValueError as e:
    print(f"ValueError: {e}")

try:
    check_password_strength(12345678)
except TypeError as e:
    print(f"TypeError: {e}")
```

<script>
(function () {
  var labId = 'ch05-lab';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';
    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var codeMirror = cell.querySelector('.CodeMirror');
    if (codeMirror && codeMirror.CodeMirror && typeof codeMirror.CodeMirror.getValue === 'function') {
      return codeMirror.CodeMirror.getValue();
    }
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent) {
      return Array.from(cmContent.querySelectorAll('.cm-line')).map(function (line) { return line.textContent; }).join('\n');
    }
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.from(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 1)] = codeFromCell(cell);
    });
    return code;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch05-lab-feedback-list');
    if (list) list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row ? row.querySelector('span') : null;
      var item = feedback ? feedback[question] : null;
      if (!row || !message || !item) return;
      row.classList.remove('correct', 'incorrect');
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch05-lab-feedback-list');
    if (list) list.hidden = true;
    grader.querySelectorAll('.ch05-lab-feedback-row').forEach(function (row) {
      row.classList.remove('correct', 'incorrect');
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var saved = text ? JSON.parse(text) : null;
        if (!response.ok || !saved || !saved.ok) throw new Error(saved && (saved.message || saved.error) ? (saved.message || saved.error) : 'Submission API is unavailable.');
        return saved;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch05-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var result = document.getElementById('ch05-lab-result');
    var submit = document.getElementById('ch05-lab-submit');
    var actions = grader.querySelector('.ch05-lab-actions');
    var reset = document.getElementById('ch05-lab-reset');

    loadSessionIdentity(identity, actions, submit, result);

    submit.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = 'Submitting code.';
      submit.disabled = true;
      saveAttempt({
        lab_id: labId,
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          if (payload.feedback) applyFeedback(grader, payload.feedback);
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
</script>
