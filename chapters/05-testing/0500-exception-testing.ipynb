# Exceptions & Testing

Errors are inevitable in any program. This chapter teaches you to recover from runtime failures and find logic errors before they reach users.

- Exception handling for runtime failures
- Tracebacks as debugging evidence
- Custom exceptions for clearer error reporting
- Tests as executable checks of program behavior

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Exception | An object that represents an error or unusual condition during program execution. |
| 2 | Exception handling | Code that responds to exceptions so a program can recover or fail clearly. |
| 3 | `try` block | The block of code where Python watches for exceptions. |
| 4 | `except` block | The block that runs when a matching exception occurs. |
| 5 | `else` block | The block that runs after a `try` block if no exception occurs. |
| 6 | `finally` block | The block that runs whether or not an exception occurs. |
| 7 | Traceback | A report showing the call path and location of an exception. |
| 8 | Debugging | The process of finding and fixing problems in a program. |
| 9 | Runtime error | An error that occurs while a program is running. |
| 10 | Semantic error | A logic error where code runs but produces the wrong result. |
| 11 | Raise | To deliberately signal an exception with the `raise` statement. |
| 12 | Custom exception | An exception class defined for a program-specific error case. |
| 13 | Logging | Recording events, warnings, and errors from a running program. |
| 14 | Log level | A severity category for a log message, such as DEBUG, INFO, WARNING, ERROR, or CRITICAL. |
| 15 | Unit test | A test that checks a small part of a program in isolation. |
| 16 | Assertion | A check that a condition is true during a test. |
| 17 | Doctest | A test written in a docstring using interactive Python examples. |
| 18 | Fixture | Setup data or resources used by tests. |
| 19 | Mock | A replacement object used to isolate code from an external dependency. |
| 20 | Test coverage | A measure of which lines or branches of code were exercised by tests. |
