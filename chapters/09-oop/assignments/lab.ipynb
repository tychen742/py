```python
### Question 1: MoneyMarketAccount subclass
# BankAccount is defined below. Define MoneyMarketAccount as a
# subclass of BankAccount with __init__(self, owner, balance=0.0,
# rate=0.02) that calls super().__init__(owner, balance) and stores
# self.rate.
# Add a method apply_interest(self) that increases self.balance by
# self.balance * self.rate.
# Create account = MoneyMarketAccount("Priya", 5000, 0.02).
# Methods are values too: store the bound method in a variable named
# apply_interest, then call apply_interest() through that variable.
# Print exactly these two lines:
# Owner: Priya
# Balance after interest: 5100.00
class BankAccount:
    def __init__(self, owner, balance=0.0):
        self.owner = owner
        self.balance = balance
### Your code starts here.



### Your code ends here.
```

```python
### Solution
class BankAccount:
    def __init__(self, owner, balance=0.0):
        self.owner = owner
        self.balance = balance

class MoneyMarketAccount(BankAccount):
    def __init__(self, owner, balance=0.0, rate=0.02):
        super().__init__(owner, balance)
        self.rate = rate

    def apply_interest(self):
        self.balance += self.balance * self.rate


account = MoneyMarketAccount("Priya", 5000, 0.02)
apply_interest = account.apply_interest
apply_interest()

print(f"Owner: {account.owner}")
print(f"Balance after interest: {account.balance:.2f}")
```

```python
### Question 2: Readable and unambiguous account display
# Define class BankAccount with __init__(self, owner, balance=0.0).
# Add __str__ returning "owner: $balance" with balance shown with
# 2 decimal places, for example "Alice: $120.50".
# Add __repr__ returning "BankAccount('owner', balance)", for
# example "BankAccount('Alice', 120.5)".
# Create accounts, a list holding BankAccount("Alice", 120.50) and
# BankAccount("Bob", 75.0).
# Loop over accounts and print each one (this uses __str__), then
# print accounts itself (this uses __repr__ for each element).
# Print exactly these three lines:
# Alice: $120.50
# Bob: $75.00
# [BankAccount('Alice', 120.5), BankAccount('Bob', 75.0)]
### Your code starts here.



### Your code ends here.
```

```python
### Solution
class BankAccount:
    def __init__(self, owner, balance=0.0):
        self.owner = owner
        self.balance = balance

    def __str__(self):
        return f"{self.owner}: ${self.balance:.2f}"

    def __repr__(self):
        return f"BankAccount({self.owner!r}, {self.balance})"


accounts = [BankAccount("Alice", 120.50), BankAccount("Bob", 75.0)]

for account in accounts:
    print(account)

print(accounts)
```

```python
### Question 3: Balance after fee
# Define class BankAccount with a class attribute flat_fee = 5.0 and
# __init__(self, owner, balance=0.0) that stores balance in
# self._balance.
# Add a read-only property named balance that returns self._balance.
# Add a read-only property named balance_after_fee that returns
# self._balance - self.flat_fee.
# Create account = BankAccount("Sam", 250.0).
# Print exactly these two lines:
# Balance: 250.0
# Balance after fee: 245.0
### Your code starts here.



### Your code ends here.
```

```python
### Solution
class BankAccount:
    flat_fee = 5.0

    def __init__(self, owner, balance=0.0):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    @property
    def balance_after_fee(self):
        return self._balance - self.flat_fee


account = BankAccount("Sam", 250.0)

print("Balance:", account.balance)
print("Balance after fee:", account.balance_after_fee)
```

```python
### Question 4: from_string classmethod
# Define class BankAccount with __init__(self, owner, balance=0.0).
# Add a classmethod from_string(cls, owner, balance_str) that
# returns cls(owner, float(balance_str)).
# Classmethods are values too: store BankAccount.from_string in a
# variable named build, then call build("Jordan", "482.75") to
# create account.
# Print exactly these two lines:
# Owner: Jordan
# Balance: 482.75
### Your code starts here.



### Your code ends here.
```

```python
### Solution
class BankAccount:
    def __init__(self, owner, balance=0.0):
        self.owner = owner
        self.balance = balance

    @classmethod
    def from_string(cls, owner, balance_str):
        return cls(owner, float(balance_str))


build = BankAccount.from_string
account = build("Jordan", "482.75")

print(f"Owner: {account.owner}")
print(f"Balance: {account.balance}")
```

```python
### Question 5: Compare and sort accounts by balance
# Define class BankAccount with __init__(self, owner, balance=0.0).
# Add __eq__(self, other) that compares self.balance to
# other.balance.
# Add __lt__(self, other) that compares self.balance to
# other.balance.
# Create a = BankAccount("Alice", 500), b = BankAccount("Bob", 500),
# c = BankAccount("Charlie", 200), and accounts = [a, b, c].
# Build ordered by calling sorted(accounts), then owners as the list
# of owner names in that order.
# Print exactly these two lines:
# Equal balances: True
# Sorted owners: ['Charlie', 'Alice', 'Bob']
### Your code starts here.



### Your code ends here.
```

```python
### Solution
class BankAccount:
    def __init__(self, owner, balance=0.0):
        self.owner = owner
        self.balance = balance

    def __eq__(self, other):
        return self.balance == other.balance

    def __lt__(self, other):
        return self.balance < other.balance


a = BankAccount("Alice", 500)
b = BankAccount("Bob", 500)
c = BankAccount("Charlie", 200)

accounts = [a, b, c]
ordered = sorted(accounts)
owners = [account.owner for account in ordered]

print("Equal balances:", a == b)
print("Sorted owners:", owners)
```

<script>
(function () {
  var labId = 'ch09-lab';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';
    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var codeMirror = cell.querySelector('.CodeMirror');
    if (codeMirror && codeMirror.CodeMirror && typeof codeMirror.CodeMirror.getValue === 'function') {
      return codeMirror.CodeMirror.getValue();
    }
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent) {
      return Array.from(cmContent.querySelectorAll('.cm-line')).map(function (line) { return line.textContent; }).join('\n');
    }
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.from(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 1)] = codeFromCell(cell);
    });
    return code;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch09-lab-feedback-list');
    if (list) list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row ? row.querySelector('span') : null;
      var item = feedback ? feedback[question] : null;
      if (!row || !message || !item) return;
      row.classList.remove('correct', 'incorrect');
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch09-lab-feedback-list');
    if (list) list.hidden = true;
    grader.querySelectorAll('.ch09-lab-feedback-row').forEach(function (row) {
      row.classList.remove('correct', 'incorrect');
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var saved = text ? JSON.parse(text) : null;
        if (!response.ok || !saved || !saved.ok) throw new Error(saved && (saved.message || saved.error) ? (saved.message || saved.error) : 'Submission API is unavailable.');
        return saved;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch09-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var result = document.getElementById('ch09-lab-result');
    var submit = document.getElementById('ch09-lab-submit');
    var actions = grader.querySelector('.ch09-lab-actions');
    var reset = document.getElementById('ch09-lab-reset');

    loadSessionIdentity(identity, actions, submit, result);

    submit.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = 'Submitting code.';
      submit.disabled = true;
      saveAttempt({
        lab_id: labId,
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          if (payload.feedback) applyFeedback(grader, payload.feedback);
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
</script>
