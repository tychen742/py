# OOP

Object-oriented programming is a paradigm for organizing data and behavior together in reusable types.

- Classes as blueprints for objects
- Instances as concrete objects with their own state
- Methods as behavior attached to objects
- Encapsulation, inheritance, polymorphism, and abstraction
- Advanced class features for Pythonic designs

This chapter starts with foundational ideas, works through practical examples, and builds toward more advanced OOP techniques.

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Object-oriented programming | A programming style that organizes data and behavior into objects. |
| 2 | Class | A blueprint for creating objects with shared attributes and methods. |
| 3 | Instance | A specific object created from a class. |
| 4 | Attribute | Data stored on an object or class. |
| 5 | Method | A function defined inside a class and called on an object or class. |
| 6 | `self` | The conventional name for the current instance inside an instance method. |
| 7 | Constructor | Code that initializes a new object, commonly written with `__init__`. |
| 8 | Dunder method | A special method with double underscores, such as `__str__` or `__lt__`. |
| 9 | Encapsulation | Keeping related data and behavior together behind a clear interface. |
| 10 | Inheritance | Defining a class from another class to reuse or specialize behavior. |
| 11 | Polymorphism | Using a shared interface with objects of different types. |
| 12 | Abstraction | Focusing on what an object does while hiding unnecessary implementation details. |
| 13 | Method overriding | Redefining an inherited method in a child class. |
| 14 | `super` | A function used to call behavior from a parent class. |
| 15 | Dataclass | A class created with `@dataclass` to reduce boilerplate for data objects. |
| 16 | `namedtuple` | An immutable tuple-like object whose fields can be accessed by name. |
| 17 | Class variable | A variable stored on the class and shared by its instances. |
| 18 | Instance variable | A variable stored separately on each instance. |
| 19 | Static method | A method placed in a class namespace that does not receive `self` or `cls`. |
| 20 | Class method | A method that receives the class as its first argument, conventionally named `cls`. |
