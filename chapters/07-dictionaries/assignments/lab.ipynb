```python
### Question 1: Discounted price list
# A shop wants a discounted price list with uppercase item names.
# Build discounted as a dict comprehension from prices: keys are
# name.upper(), values are the price reduced by discount and rounded
# to 2 decimal places.
# Loop over discounted in sorted key order and print one line per
# item as "NAME: price". Print exactly these four lines:
# APPLE: 1.08
# BANANA: 0.45
# CHERRY: 1.8
# GRAPE: 0.72
prices = {"apple": 1.20, "banana": 0.50, "cherry": 2.00, "grape": 0.80}
discount = 0.10
### Your code starts here.



### Your code ends here.
```

```python
### Solution
prices = {"apple": 1.20, "banana": 0.50, "cherry": 2.00, "grape": 0.80}
discount = 0.10

discounted = {name.upper(): round(price * (1 - discount), 2) for name, price in prices.items()}

for name in sorted(discounted):
    print(f"{name}: {discounted[name]}")
```

```python
### Question 2: Invert the capitals dictionary
# capitals maps each state to its capital city. Build city_to_state
# by inverting capitals with a dict comprehension (city as key, state
# as value).
# Print the state for "Topeka" using city_to_state.
# Then loop over city_to_state in sorted key order and print one line
# per city as "city -> state". Print exactly these four lines:
# State for Topeka: Kansas
# Jefferson City -> Missouri
# Little Rock -> Arkansas
# Topeka -> Kansas
capitals = {"Missouri": "Jefferson City", "Kansas": "Topeka", "Arkansas": "Little Rock"}
### Your code starts here.



### Your code ends here.
```

```python
### Solution
capitals = {"Missouri": "Jefferson City", "Kansas": "Topeka", "Arkansas": "Little Rock"}

city_to_state = {city: state for state, city in capitals.items()}

print("State for Topeka:", city_to_state["Topeka"])
for city in sorted(city_to_state):
    print(f"{city} -> {city_to_state[city]}")
```

```python
### Question 3: Scores keyed by (name, subject)
# scores maps a (name, subject) tuple to a score.
# Print Alice's math score using the tuple key ("alice", "math").
# Then, for each unique name in sorted order, compute the average of
# that student's scores and print it with one decimal place.
# Print exactly these three lines:
# Alice math score: 92
# alice average: 90.0
# bob average: 78.0
scores = {
    ("alice", "math"): 92, ("alice", "science"): 88,
    ("bob", "math"): 75, ("bob", "science"): 81,
}
### Your code starts here.



### Your code ends here.
```

```python
### Solution
scores = {
    ("alice", "math"): 92, ("alice", "science"): 88,
    ("bob", "math"): 75, ("bob", "science"): 81,
}

print("Alice math score:", scores[("alice", "math")])

names = sorted({name for name, subject in scores})
for name in names:
    values = [value for (n, subject), value in scores.items() if n == name]
    average = sum(values) / len(values)
    print(f"{name} average: {average:.1f}")
```

```python
### Question 4: Validate orders against inventory
# inventory maps each SKU on hand to its quantity. order_skus lists
# SKUs a customer wants to buy.
# A dictionary lookup by key is fast because Python hashes the key
# directly to a bucket, unlike scanning a list item by item.
# Use inventory.get(sku, -1) to test membership: build valid_orders
# with the SKUs that ARE keys in inventory (get(sku, -1) != -1), and
# invalid_orders with the SKUs that are NOT keys in inventory.
# Print exactly these three lines:
# Valid orders: ['sku1', 'sku3']
# Invalid orders: ['sku4', 'sku5']
# Valid count: 2
inventory = {"sku1": 10, "sku2": 0, "sku3": 5}
order_skus = ["sku1", "sku4", "sku3", "sku5"]
### Your code starts here.



### Your code ends here.
```

```python
### Solution
inventory = {"sku1": 10, "sku2": 0, "sku3": 5}
order_skus = ["sku1", "sku4", "sku3", "sku5"]

valid_orders = [sku for sku in order_skus if inventory.get(sku, -1) != -1]
invalid_orders = [sku for sku in order_skus if inventory.get(sku, -1) == -1]

print("Valid orders:", valid_orders)
print("Invalid orders:", invalid_orders)
print("Valid count:", len(valid_orders))
```

```python
### Question 5: Update a nested student record
# student_record has a nested "grades" dictionary and a "clubs" list.
# Use .update() on student_record["grades"] to add a "history" grade
# of 78.
# Use .pop() on student_record to remove the "clubs" key and store
# the removed value in removed_clubs.
# Use .setdefault() on student_record to add an "advisor" key with
# the value "Dr. Lee" (only if it is not already present).
# Print removed_clubs, then the advisor, then each subject and grade
# in student_record["grades"] in sorted key order. Print exactly
# these five lines:
# Removed clubs: ['chess']
# Advisor: Dr. Lee
# history: 78
# math: 90
# science: 85
student_record = {
    "name": "Riya",
    "grades": {"math": 90, "science": 85},
    "clubs": ["chess"],
}
### Your code starts here.



### Your code ends here.
```

```python
### Solution
student_record = {
    "name": "Riya",
    "grades": {"math": 90, "science": 85},
    "clubs": ["chess"],
}

student_record["grades"].update({"history": 78})
removed_clubs = student_record.pop("clubs")
student_record.setdefault("advisor", "Dr. Lee")

print("Removed clubs:", removed_clubs)
print("Advisor:", student_record["advisor"])
for subject in sorted(student_record["grades"]):
    print(f"{subject}: {student_record['grades'][subject]}")
```

<script>
(function () {
  var labId = 'ch07-lab';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';
    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var codeMirror = cell.querySelector('.CodeMirror');
    if (codeMirror && codeMirror.CodeMirror && typeof codeMirror.CodeMirror.getValue === 'function') {
      return codeMirror.CodeMirror.getValue();
    }
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent) {
      return Array.from(cmContent.querySelectorAll('.cm-line')).map(function (line) { return line.textContent; }).join('\n');
    }
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.from(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 1)] = codeFromCell(cell);
    });
    return code;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch07-lab-feedback-list');
    if (list) list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row ? row.querySelector('span') : null;
      var item = feedback ? feedback[question] : null;
      if (!row || !message || !item) return;
      row.classList.remove('correct', 'incorrect');
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch07-lab-feedback-list');
    if (list) list.hidden = true;
    grader.querySelectorAll('.ch07-lab-feedback-row').forEach(function (row) {
      row.classList.remove('correct', 'incorrect');
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var saved = text ? JSON.parse(text) : null;
        if (!response.ok || !saved || !saved.ok) throw new Error(saved && (saved.message || saved.error) ? (saved.message || saved.error) : 'Submission API is unavailable.');
        return saved;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch07-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var result = document.getElementById('ch07-lab-result');
    var submit = document.getElementById('ch07-lab-submit');
    var actions = grader.querySelector('.ch07-lab-actions');
    var reset = document.getElementById('ch07-lab-reset');

    loadSessionIdentity(identity, actions, submit, result);

    submit.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = 'Submitting code.';
      submit.disabled = true;
      saveAttempt({
        lab_id: labId,
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          if (payload.feedback) applyFeedback(grader, payload.feedback);
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
</script>
