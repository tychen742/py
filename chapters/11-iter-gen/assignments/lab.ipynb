```python
### Question 1: Custom Iterator: Threshold Filter
# A sales team wants to review only the transactions above a threshold.
# Implement a class AboveThreshold(values, threshold) with __iter__/__next__
# that returns, one at a time, only the values greater than threshold.
# Skip values at or below threshold instead of stopping on them.
# Raise StopIteration once every value has been checked.
# Collect the filtered values with list(), then print exactly these two lines:
# Sales above 1000: [1200, 2100, 1600]
# Count above 1000: 3
sales = [1200, 850, 2100, 430, 975, 1600]
### Your code starts here.



### Your code ends here.
```

```python
### Solution
sales = [1200, 850, 2100, 430, 975, 1600]

class AboveThreshold:
    def __init__(self, values, threshold):
        self.values = list(values)
        self.threshold = threshold
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        while self.index < len(self.values):
            value = self.values[self.index]
            self.index += 1
            if value > self.threshold:
                return value
        raise StopIteration

big_sales = AboveThreshold(sales, 1000)
result = list(big_sales)
print("Sales above 1000:", result)
print("Count above 1000:", len(result))
```

```python
### Question 2: Custom Iterator: Batch Sales Data
# Large sales exports are processed in fixed-size batches.
# Implement a class Batched(values, size) with __iter__/__next__ that
# returns successive lists (batches) of length size from values.
# The final batch may be shorter than size if values does not divide evenly.
# Raise StopIteration once every value has been placed in a batch.
# Loop over Batched(sales, 2) with a for loop and print each batch (one per line):
# [1200, 850]
# [2100, 430]
# [975, 1600]
sales = [1200, 850, 2100, 430, 975, 1600]
### Your code starts here.



### Your code ends here.
```

```python
### Solution
sales = [1200, 850, 2100, 430, 975, 1600]

class Batched:
    def __init__(self, values, size):
        self.values = list(values)
        self.size = size
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.values):
            raise StopIteration
        batch = self.values[self.index:self.index + self.size]
        self.index += self.size
        return batch

for batch in Batched(sales, 2):
    print(batch)
```

```python
### Question 3: Generator Function: Running Maximum
# Write a generator function running_max(numbers) that yields the
# largest value seen so far after each element (a running maximum).
# Example: running_max([1, 5, 3, 8]) yields 1, 5, 5, 8.
# Collect the results with list() on the sales data and print exactly:
# [1200, 1200, 2100, 2100, 2100, 2100]
sales = [1200, 850, 2100, 430, 975, 1600]
### Your code starts here.



### Your code ends here.
```

```python
### Solution
sales = [1200, 850, 2100, 430, 975, 1600]

def running_max(numbers):
    current_max = None
    for n in numbers:
        if current_max == None or n > current_max:
            current_max = n
        yield current_max

print(list(running_max(sales)))
```

```python
### Question 4: Generator Expressions with islice, any, and all
# Use a generator expression to compute a 10% discount on each sale:
# round(s * 0.9, 2) for each s in sales.
# Use itertools.islice to take only the first 3 discounted values.
# Use any() with a generator expression to check if any sale exceeds 1000.
# Use all() with a generator expression to check if every sale exceeds 400.
# Print exactly these three lines:
# First three discounted: [1080.0, 765.0, 1890.0]
# Any sale over 1000: True
# All sales over 400: True
import itertools

sales = [1200, 850, 2100, 430, 975, 1600]
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import itertools

sales = [1200, 850, 2100, 430, 975, 1600]

discounted_gen = (round(s * 0.9, 2) for s in sales)
first_three = list(itertools.islice(discounted_gen, 3))
any_over_1000 = any(s > 1000 for s in sales)
all_over_400 = all(s > 400 for s in sales)

print("First three discounted:", first_three)
print("Any sale over 1000:", any_over_1000)
print("All sales over 400:", all_over_400)
```

```python
### Question 5: Infinite Generator with itertools.islice
# Write a generator function grow_sales(base, rate) that models
# compound growth: it yields base, then repeatedly multiplies the
# previous value by (1 + rate) and yields the result, forever (while True).
# Round each yielded value to 2 decimal places.
# Always bound consumption of an infinite generator with itertools.islice
# so the loop terminates.
# Take the first 5 values from grow_sales(1000, 0.1) and print exactly:
# Projected sales: [1000, 1100.0, 1210.0, 1331.0, 1464.1]
import itertools
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import itertools

def grow_sales(base, rate):
    value = base
    while True:
        yield value
        value = round(value * (1 + rate), 2)

first_5 = list(itertools.islice(grow_sales(1000, 0.1), 5))
print("Projected sales:", first_5)
```

<script>
(function () {
  var labId = 'ch11-lab';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';
    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var codeMirror = cell.querySelector('.CodeMirror');
    if (codeMirror && codeMirror.CodeMirror && typeof codeMirror.CodeMirror.getValue === 'function') {
      return codeMirror.CodeMirror.getValue();
    }
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent) {
      return Array.from(cmContent.querySelectorAll('.cm-line')).map(function (line) { return line.textContent; }).join('\n');
    }
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.from(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 1)] = codeFromCell(cell);
    });
    return code;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch11-lab-feedback-list');
    if (list) list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row ? row.querySelector('span') : null;
      var item = feedback ? feedback[question] : null;
      if (!row || !message || !item) return;
      row.classList.remove('correct', 'incorrect');
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch11-lab-feedback-list');
    if (list) list.hidden = true;
    grader.querySelectorAll('.ch11-lab-feedback-row').forEach(function (row) {
      row.classList.remove('correct', 'incorrect');
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var saved = text ? JSON.parse(text) : null;
        if (!response.ok || !saved || !saved.ok) throw new Error(saved && (saved.message || saved.error) ? (saved.message || saved.error) : 'Submission API is unavailable.');
        return saved;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch11-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var result = document.getElementById('ch11-lab-result');
    var submit = document.getElementById('ch11-lab-submit');
    var actions = grader.querySelector('.ch11-lab-actions');
    var reset = document.getElementById('ch11-lab-reset');

    loadSessionIdentity(identity, actions, submit, result);

    submit.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = 'Submitting code.';
      submit.disabled = true;
      saveAttempt({
        lab_id: labId,
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          if (payload.feedback) applyFeedback(grader, payload.feedback);
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
</script>
