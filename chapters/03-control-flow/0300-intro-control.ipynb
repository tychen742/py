# Control Flow

Control flow lets a program decide what to do next. This chapter introduces the branching and looping structures that turn straight-line code into programs that respond to data.

- Conditional statements with `if`, `elif`, and `else`
- `for` loops for iterating over data
- `while` loops for condition-based repetition
- Loop-control keywords for special cases

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Control flow | The order in which a program runs statements. |
| 2 | Conditional statement | A statement that chooses whether to run code based on a condition. |
| 3 | Boolean expression | An expression that evaluates to `True` or `False`. |
| 4 | Branch | One possible path through a conditional statement. |
| 5 | `if` statement | A conditional statement that runs a block only when its condition is true. |
| 6 | `else` clause | The branch that runs when the `if` condition is false. |
| 7 | `elif` clause | A chained conditional branch that tests another condition. |
| 8 | Nested conditional | A conditional statement placed inside another conditional statement. |
| 9 | Conditional expression | A compact expression that chooses between two values. |
| 10 | Iteration | Repeating a block of code. |
| 11 | Iterable | An object that can provide its items one at a time. |
| 12 | Iterator | An object that tracks its position while producing items one at a time. |
| 13 | `for` loop | A loop that repeats once for each item in an iterable. |
| 14 | `range` | A built-in type commonly used to generate integer sequences for loops. |
| 15 | `while` loop | A loop that repeats while a condition remains true. |
| 16 | Sentinel | A special value that signals when a data-controlled loop should stop. |
| 17 | Infinite loop | A loop whose condition never becomes false unless something interrupts it. |
| 18 | `break` statement | A statement that exits the nearest enclosing loop immediately. |
| 19 | `continue` statement | A statement that skips the rest of the current loop iteration. |
| 20 | Running total | A variable updated inside a loop to accumulate a total such as total rainfall. |
