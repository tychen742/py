# Introduction

This chapter gives you a high-level view of what programming is, how Python fits into the broader computing landscape, and which ideas you will rely on throughout the book.

- Programming as a way to express steps and decisions
- Python as a readable, practical language
- Jupyter as a place to experiment with code
- Basic syntax as the notation for programming ideas

Python is a practical first language because it lets you focus on problem solving without fighting the syntax too early. Its readability, standard library, and ecosystem also make it useful for automation, scripting, data analysis, and software development.
This chapter has two sections: programming concepts first, then basic Python syntax.

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Program | A set of instructions written for a computer to execute. |
| 2 | Programming language | A formal notation for writing instructions that a computer can interpret or compile. |
| 3 | Natural language | A human language, such as English, that allows ambiguity and depends on context. |
| 4 | Formal language | A precisely defined language with strict syntax and meaning. |
| 5 | Abstraction | A simplified view that hides lower-level details so you can focus on the idea that matters. |
| 6 | Interpreted language | A language commonly run by executing source code through an interpreter. |
| 7 | Compiled language | A language commonly translated into another form before the program runs. |
| 8 | Programming construct | A basic building block used to express computation, such as a statement or expression. |
| 9 | Expression | A piece of code that produces a value. |
| 10 | Statement | A complete instruction that Python can execute. |
| 11 | Number system | A way to represent numbers using a base and a set of digits. |
| 12 | Binary | A base-2 number system that uses only 0 and 1. |
| 13 | Hexadecimal | A base-16 number system often used as a compact way to write binary values. |
| 14 | Character encoding | A mapping between characters and numeric codes. |
| 15 | Syntax | The formal rules for writing valid statements in a programming language. |
| 16 | Indentation | Leading whitespace that Python uses to show block structure. |
| 17 | Block | A group of statements that belong together under a control structure or definition. |
| 18 | `print` | A built-in function used to display output. |
| 19 | F-string | A formatted string literal that can include Python expressions inside braces. |
| 20 | `input` | A built-in function that reads text typed by the user. |
