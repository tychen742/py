<script>
(function () {
  var quizId = 'ch01-preview';
  var questions = [
    { id: 'q1', prompt: 'Which term means a set of instructions that a computer can execute?', choices: { A: 'Program', B: 'Comment', C: 'Expression', D: 'Jupyter notebook' }, answer: 'A' },
    { id: 'q2', prompt: 'Which term means the process of writing precise instructions for a computer to follow?', choices: { A: 'Syntax', B: 'Programming', C: 'Source code', D: 'Statement' }, answer: 'B' },
    { id: 'q3', prompt: 'Which term names a readable, general-purpose language used for automation, data analysis, web development, and many other tasks?', choices: { A: 'Interpreter', B: 'Algorithm', C: 'Python', D: 'Function' }, answer: 'C' },
    { id: 'q4', prompt: 'Which term means a program that runs source code by translating and executing it, often one statement at a time?', choices: { A: 'Function', B: 'Interpreter', C: 'Comment', D: 'Expression' }, answer: 'B' },
    { id: 'q5', prompt: 'Which term means human-readable instructions written in a programming language?', choices: { A: 'Statement', B: 'Jupyter notebook', C: 'Syntax', D: 'Source code' }, answer: 'D' },
    { id: 'q6', prompt: 'What does this code display? print("red", "blue", sep="-")', choices: { A: 'red blue', B: 'red-blue', C: 'red, blue', D: 'sep=red-blue' }, answer: 'B' },
    { id: 'q7', prompt: 'What does this code display? name = "Ada"; print(f"Hello, {name}!")', choices: { A: 'Hello, name!', B: 'Hello, {name}!', C: 'Hello, Ada!', D: 'f"Hello, Ada!"' }, answer: 'C' },
    { id: 'q8', prompt: 'Which statement about input() is correct?', choices: { A: 'input() returns an integer by default.', B: 'input() returns a string by default.', C: 'input() prints output without waiting.', D: 'input() only works inside loops.' }, answer: 'B' },
    { id: 'q9', prompt: 'In Python, what does a line that starts with # usually do?', choices: { A: 'Creates a variable.', B: 'Starts a comment ignored by Python.', C: 'Imports a module.', D: 'Ends the program.' }, answer: 'B' },
    { id: 'q10', prompt: 'What does this code display? item_number = 64; print(hex(item_number))', choices: { A: '64', B: '0b1000000', C: '0x40', D: '0o100' }, answer: 'C' }
  ];

  function text(value) {
    return document.createTextNode(value);
  }

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || '',
      student_identifier: params.get('student_identifier') || params.get('student_id') || params.get('email') || ''
    };
  }

  function loadSessionIdentity(identity, identifierInput, identityNote) {
    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || identity.student_identifier || '';
        identifierInput.value = identity.student_identifier || identity.canvas_user_id || '';
        identifierInput.readOnly = true;
        identityNote.textContent = payload.identity.display_name ? 'Signed in from Canvas as ' + payload.identity.display_name + '.' : 'Signed in from Canvas.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function renderQuestions(container) {
    questions.forEach(function (question, index) {
      var card = document.createElement('div');
      card.className = 'ch01-preview-question';
      card.dataset.question = question.id;

      var prompt = document.createElement('p');
      prompt.appendChild(text((index + 1) + '. ' + question.prompt));
      card.appendChild(prompt);

      Object.keys(question.choices).forEach(function (letter) {
        var label = document.createElement('label');
        var input = document.createElement('input');
        input.type = 'radio';
        input.name = question.id;
        input.value = letter;
        label.appendChild(input);
        label.appendChild(text(letter + '. ' + question.choices[letter]));
        card.appendChild(label);
      });

      var feedback = document.createElement('div');
      feedback.className = 'ch01-preview-feedback';
      feedback.setAttribute('aria-live', 'polite');
      card.appendChild(feedback);
      container.appendChild(card);
    });
  }

  function collectAnswers(quiz) {
    var answers = {};
    questions.forEach(function (question) {
      var selected = quiz.querySelector('input[name=' + question.id + ']:checked');
      answers[question.id] = selected ? selected.value : '';
    });
    return answers;
  }

  function gradeAnswers(answers) {
    var feedback = {};
    var score = 0;
    questions.forEach(function (question) {
      var submitted = answers[question.id] || '';
      var correct = submitted === question.answer;
      if (correct) score += 1;
      feedback[question.id] = {
        submitted: submitted !== '',
        correct: correct,
        message: submitted === '' ? 'No answer submitted.' : (correct ? 'Correct.' : 'Try again.')
      };
    });
    return { ok: true, score: score, max_score: questions.length, feedback: feedback, canvas_sync_status: 'not configured' };
  }

  function applyFeedback(quiz, feedback) {
    questions.forEach(function (question) {
      var card = quiz.querySelector('[data-question=' + question.id + ']');
      var message = card.querySelector('.ch01-preview-feedback');
      var item = feedback[question.id];
      message.className = 'ch01-preview-feedback';
      if (!item || !item.submitted) {
        message.textContent = item && item.message ? item.message : 'No answer submitted.';
        message.classList.add('incorrect');
      } else if (item.correct) {
        message.textContent = item.message || 'Correct.';
        message.classList.add('correct');
      } else {
        message.textContent = item.message || 'Try again.';
        message.classList.add('incorrect');
      }
    });
  }

  function saveBestScore(payload) {
    try {
      var key = quizId + ':best-score';
      var best = Number(localStorage.getItem(key) || '0');
      if (payload.score > best) localStorage.setItem(key, String(payload.score));
      return Math.max(best, payload.score);
    } catch (error) {
      return payload.score;
    }
  }

  function submitAttempt(identity, identifierInput, answers) {
    return fetch('/api/v1/quiz-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify({
        quiz_id: quizId,
        answers: answers,
        student_identifier: identifierInput.value,
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
    })
      .then(function (response) {
        return response.text().then(function (body) {
          var payload = null;
          try {
            payload = body ? JSON.parse(body) : null;
          } catch (error) {
            throw new Error('Server returned a non-JSON response.');
          }
          if (!payload || !response.ok || !payload.ok) {
            throw new Error(payload && (payload.message || payload.error) ? (payload.message || payload.error) : 'Submission API is unavailable.');
          }
          return payload;
        });
      });
  }

  function initPreviewQuiz() {
    var quiz = document.getElementById('ch01-preview-quiz');
    if (!quiz || quiz.dataset.ready === 'true') return;
    quiz.dataset.ready = 'true';

    var identity = getParams();
    var identifierInput = document.getElementById('ch01-student-identifier');
    var questionsContainer = document.getElementById('ch01-preview-questions');
    var identityNote = document.getElementById('ch01-preview-identity-note');
    var result = document.getElementById('ch01-preview-result');
    var submit = document.getElementById('ch01-preview-submit');
    var reset = document.getElementById('ch01-preview-reset');

    identifierInput.value = identity.student_identifier || identity.canvas_user_id || '';
    renderQuestions(questionsContainer);
    loadSessionIdentity(identity, identifierInput, identityNote);

    submit.addEventListener('click', function () {
      var answers = collectAnswers(quiz);
      var localPayload = gradeAnswers(answers);
      submit.disabled = true;
      result.textContent = 'Submitting...';

      submitAttempt(identity, identifierInput, answers)
        .then(function (payload) {
          saveBestScore(payload);
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          applyFeedback(quiz, payload.feedback || localPayload.feedback || {});
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + (payload.canvas_sync_status || 'not configured') + '.';
        })
        .catch(function (error) {
          applyFeedback(quiz, localPayload.feedback || {});
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      quiz.querySelectorAll('input[type=radio]').forEach(function (input) {
        input.checked = false;
      });
      quiz.querySelectorAll('.ch01-preview-feedback').forEach(function (feedback) {
        feedback.className = 'ch01-preview-feedback';
        feedback.textContent = '';
      });
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initPreviewQuiz);
  } else {
    initPreviewQuiz();
  }
})();
</script>
