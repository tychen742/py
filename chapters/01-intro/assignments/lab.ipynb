```python
### Question 2: Expressions and assignment statements
# The variables below describe a short reading plan.
# Write assignment statements that calculate total_pages and average_pages_per_day.
# Print both values. Use round(average_pages_per_day, 2) for the average.
hours = 6
pages_per_hour = 14
review_pages = 9
days = 5
### Your code starts here.



### Your code ends here.
```

```python
### Solution
hours = 6
pages_per_hour = 14
review_pages = 9
days = 5

total_pages = hours * pages_per_hour + review_pages
average_pages_per_day = total_pages / days

print("Total pages:", total_pages)
print("Average per day:", round(average_pages_per_day, 2))
```

```python
### Question 4: Convert text input to numbers
# input() returns text, so numeric input usually needs conversion.
# The variable below simulates text a user typed into input().
# Convert minutes_text to an integer named total_minutes.
# Calculate full hours in a variable named hours_part.
# Calculate leftover minutes in a variable named minutes_part.
# Print hours_part and minutes_part.
minutes_text = "135"
### Your code starts here.



### Your code ends here.
```

```python
### Solution
minutes_text = "135"

total_minutes = int(minutes_text)
hours_part = total_minutes // 60
minutes_part = total_minutes % 60

print("Hours:", hours_part)
print("Minutes:", minutes_part)
```

```python
### Question 5: Number systems and character encoding
# A simple inventory system uses product code "C" and item number 64.
# Save the decimal code for "C" in c_decimal using ord().
# Save the binary representation of c_decimal in c_binary using bin().
# Save the hexadecimal representation of item_number in item_hex using hex().
# Print c_decimal, c_binary, and item_hex.
item_number = 64
### Your code starts here.



### Your code ends here.
```

```python
### Solution
item_number = 64

c_decimal = ord("C")
c_binary = bin(c_decimal)
item_hex = hex(item_number)

print("C decimal:", c_decimal)
print("C binary:", c_binary)
print("Item hex:", item_hex)
```

<script>
(function () {
  var labId = 'ch01-lab';
  var answerKeys = ['first_line', 'second_line', 'total_pages', 'average_pages_per_day', 'summary_line', 'sep_line', 'total_minutes', 'hours_part', 'minutes_part', 'c_decimal', 'c_binary', 'item_hex'];
  var checks = {
    q1: { keys: ['first_line', 'second_line'], answers: { first_line: 'ThinkPy Chapter 1', second_line: 'Python is running.' } },
    q2: { keys: ['total_pages', 'average_pages_per_day'], answers: { total_pages: '93', average_pages_per_day: '18.6' } },
    q3: { keys: ['summary_line', 'sep_line'], answers: { summary_line: 'sales: 128 rows x 5 columns', sep_line: 'sales|128|5' } },
    q4: { keys: ['total_minutes', 'hours_part', 'minutes_part'], answers: { total_minutes: '135', hours_part: '2', minutes_part: '15' } },
    q5: { keys: ['c_decimal', 'c_binary', 'item_hex'], answers: { c_decimal: '67', c_binary: '0b1000011', item_hex: '0x40' } }
  };

  function normalize(value) {
    return String(value || '').trim();
  }

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || '',
      student_identifier: params.get('student_identifier') || params.get('student_id') || params.get('email') || ''
    };
  }

  function loadSessionIdentity(identity, identifierInput, identityNote) {
    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || identity.student_identifier || '';
        identifierInput.value = identity.student_identifier || identity.canvas_user_id || '';
        identifierInput.readOnly = true;
        identityNote.textContent = payload.identity.display_name ? 'Signed in from Canvas as ' + payload.identity.display_name + '.' : 'Signed in from Canvas.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function collectAnswers(grader) {
    var answers = {};
    answerKeys.forEach(function (key) {
      var input = grader.querySelector('[data-answer=' + key + ']');
      answers[key] = input ? input.value : '';
    });
    return answers;
  }

  function gradeAnswers(answers) {
    var feedback = {};
    var score = 0;
    Object.keys(checks).forEach(function (question) {
      var check = checks[question];
      var correctCount = 0;
      check.keys.forEach(function (key) {
        if (normalize(answers[key]) === check.answers[key]) correctCount += 1;
      });
      var maxScore = 2;
      var itemScore = correctCount === check.keys.length ? maxScore : 0;
      score += itemScore;
      feedback[question] = {
        correct: itemScore === maxScore,
        score: itemScore,
        max_score: maxScore,
        message: itemScore === maxScore ? 'Correct' : 'Check the requested variable values or printed output.'
      };
    });
    return { score: score, max_score: 10, feedback: feedback };
  }

  function applyFeedback(grader, feedback) {
    Object.keys(checks).forEach(function (question) {
      var fieldset = grader.querySelector('[data-question=' + question + ']');
      var message = fieldset.querySelector('.ch01-lab-feedback');
      var item = feedback[question];
      message.className = 'ch01-lab-feedback';
      if (!item) return;
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      message.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var saved = text ? JSON.parse(text) : null;
        if (!response.ok || !saved || !saved.ok) throw new Error('not saved');
        return saved;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch01-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var identifierInput = document.getElementById('ch01-lab-student-identifier');
    var identityNote = document.getElementById('ch01-lab-identity-note');
    var result = document.getElementById('ch01-lab-result');
    var submit = document.getElementById('ch01-lab-submit');
    var reset = document.getElementById('ch01-lab-reset');

    identifierInput.value = identity.student_identifier || identity.canvas_user_id || '';
    loadSessionIdentity(identity, identifierInput, identityNote);

    submit.addEventListener('click', function () {
      var answers = collectAnswers(grader);
      var graded = gradeAnswers(answers);
      applyFeedback(grader, graded.feedback);
      result.textContent = 'Score: ' + graded.score + ' / ' + graded.max_score + '.';

      submit.disabled = true;
      saveAttempt({
        lab_id: labId,
        answers: answers,
        local_score: graded.score,
        local_max_score: graded.max_score,
        student_identifier: identifierInput.value,
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          if (payload.feedback) applyFeedback(grader, payload.feedback);
          if (typeof payload.score !== 'undefined') {
            result.textContent = 'Score: ' + payload.score + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
          } else {
            result.textContent = 'Score: ' + graded.score + ' / ' + graded.max_score + '. Attempt saved.';
          }
        })
        .catch(function () {
          result.textContent = 'Score: ' + graded.score + ' / ' + graded.max_score + '. Attempt checked locally; API save is unavailable.';
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      grader.querySelectorAll('input[data-answer]').forEach(function (input) { input.value = ''; });
      grader.querySelectorAll('.ch01-lab-feedback').forEach(function (feedback) { feedback.className = 'ch01-lab-feedback'; feedback.textContent = ''; });
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
</script>
