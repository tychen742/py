# Introduction

This chapter gives you a high-level view of what programming is, how Python fits into the broader computing landscape, and which ideas you will rely on throughout the book.

- Programming as a way to express steps and decisions
- Python as a readable, practical language
- Jupyter as a place to experiment with code
- Basic syntax as the notation for programming ideas

Python is a practical first language because it lets you focus on problem solving without fighting the syntax too early. Its readability, standard library, and ecosystem also make it useful for automation, scripting, data analysis, and software development.
This chapter has two sections: programming concepts first, then basic Python syntax.

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Program | A set of instructions that a computer can execute. |
| 2 | Programming | The process of writing precise instructions for a computer to follow. |
| 3 | Python | A readable, general-purpose programming language used for automation, data analysis, web development, and many other tasks. |
| 4 | Interpreter | A program that runs source code by translating and executing it, often one statement at a time. |
| 5 | Source code | Human-readable instructions written in a programming language. |
| 6 | Algorithm | A step-by-step procedure for solving a problem or completing a task. |
| 7 | Syntax | The formal rules for writing valid statements in a programming language. |
| 8 | Statement | A complete instruction that Python can execute. |
| 9 | Expression | A piece of code that produces a value. |
| 10 | Comment | Text in source code that is written for people and ignored by Python. |
| 11 | Function | A named, reusable block of code that performs a task. |
| 12 | Jupyter notebook | An interactive document that combines code, output, formatted text, and media. |
