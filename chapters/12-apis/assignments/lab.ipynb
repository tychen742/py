```python
### Question 1: Build a URL-Encoded Query String
# Build the query string for a search request without using requests
# or any network call -- just plain string and dict manipulation.
# Write a function url_encode(value) that converts value to a string
# and replaces spaces with %20.
# Build query_string by joining 'key=value' pairs from params with '&',
# url-encoding each value.
# Combine base_url, '?', and query_string into full_url and print it:
# https://api.example.test/search?q=python%20generators&page=3&limit=10
base_url = "https://api.example.test/search"
params = {"q": "python generators", "page": 3, "limit": 10}
### Your code starts here.



### Your code ends here.
```

```python
### Solution
base_url = "https://api.example.test/search"
params = {"q": "python generators", "page": 3, "limit": 10}

def url_encode(value):
    return str(value).replace(" ", "%20")

query_parts = [f"{key}={url_encode(value)}" for key, value in params.items()]
query_string = "&".join(query_parts)
full_url = f"{base_url}?{query_string}"
print(full_url)
```

```python
### Question 2: Safe Nested JSON Traversal
# payload represents a pre-fetched API response (a plain dict).
# Write a function get_city(payload) that safely returns
# payload["user"]["address"]["city"], using .get() chains with
# default {} for missing dicts and default "unknown" for the city,
# so it never raises KeyError even when fields are missing.
# Call it on payload, then on {'user': {}}, and print exactly:
# Austin
# unknown
payload = {
    "user": {
        "id": 55,
        "address": {"city": "Austin", "zip": "78701"},
    },
}
### Your code starts here.



### Your code ends here.
```

```python
### Solution
payload = {
    "user": {
        "id": 55,
        "address": {"city": "Austin", "zip": "78701"},
    },
}

def get_city(payload):
    return payload.get("user", {}).get("address", {}).get("city", "unknown")

print(get_city(payload))
print(get_city({"user": {}}))
```

```python
### Question 3: Summarize a Response Payload
# order_payload represents a pre-fetched order response.
# Write a function summarize_order(payload) that returns a dict with:
# - "order_id": payload["order_id"]
# - "total": the sum of each item's "price", minus payload.get("discount", 0),
#   rounded to 2 decimal places
# - "item_count": the number of items
# Note: payload may not include a discount key -- default it to 0.
# Call summarize_order(order_payload) and print exactly:
# {'order_id': 4821, 'total': 21.25, 'item_count': 2}
order_payload = {
    "order_id": 4821,
    "items": [{"name": "notebook", "price": 12.5}, {"name": "pen", "price": 8.75}],
}
### Your code starts here.



### Your code ends here.
```

```python
### Solution
order_payload = {
    "order_id": 4821,
    "items": [{"name": "notebook", "price": 12.5}, {"name": "pen", "price": 8.75}],
}

def summarize_order(payload):
    total = sum(item["price"] for item in payload.get("items", []))
    discount = payload.get("discount", 0)
    return {
        "order_id": payload["order_id"],
        "total": round(total - discount, 2),
        "item_count": len(payload.get("items", [])),
    }

print(summarize_order(order_payload))
```

```python
### Question 4: Simulate Pagination
# pages simulates paginated API responses already fetched into memory:
# a list of page dicts, each with 'items' and a 'next' index (or None
# on the last page -- the sentinel that stops pagination).
# Write a function fetch_all(pages) that starts at index 0, collects
# every page's 'items' into one list, and follows 'next' until it
# equals None. Return the combined list.
# Call fetch_all(pages) and print exactly:
# ['alpha', 'beta', 'gamma', 'delta', 'epsilon']
pages = [
    {"items": ["alpha", "beta"], "next": 1},
    {"items": ["gamma"], "next": 2},
    {"items": ["delta", "epsilon"], "next": None},
]
### Your code starts here.



### Your code ends here.
```

```python
### Solution
pages = [
    {"items": ["alpha", "beta"], "next": 1},
    {"items": ["gamma"], "next": 2},
    {"items": ["delta", "epsilon"], "next": None},
]

def fetch_all(pages):
    collected = []
    index = 0
    while index != None:
        page = pages[index]
        collected.extend(page["items"])
        index = page["next"]
    return collected

print(fetch_all(pages))
```

```python
### Question 5: Parse and Re-serialize a JSON Payload
# raw_response simulates a JSON string already received from an API
# (no network call -- it is just a string literal here).
# Use json.loads to parse it into parsed.
# Write a function get_result_names(parsed) that returns a list of each
# result's "name", using .get("name", "unnamed") as a safe default.
# Build summary as a JSON string (json.dumps) of a dict with
# "status" (default "unknown" if missing) and "count" (len(names)).
# Print names, then print summary, exactly:
# ['Alpha', 'Beta']
# {"status": "ok", "count": 2}
import json

raw_response = '{"status": "ok", "results": [{"id": 1, "name": "Alpha"}, {"id": 2, "name": "Beta"}]}'
parsed = json.loads(raw_response)
### Your code starts here.



### Your code ends here.
```

```python
### Solution
import json

raw_response = '{"status": "ok", "results": [{"id": 1, "name": "Alpha"}, {"id": 2, "name": "Beta"}]}'
parsed = json.loads(raw_response)

def get_result_names(parsed):
    return [item.get("name", "unnamed") for item in parsed.get("results", [])]

names = get_result_names(parsed)
summary = json.dumps({"status": parsed.get("status", "unknown"), "count": len(names)})
print(names)
print(summary)
```

<script>
(function () {
  var labId = 'ch12-lab';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';
    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var codeMirror = cell.querySelector('.CodeMirror');
    if (codeMirror && codeMirror.CodeMirror && typeof codeMirror.CodeMirror.getValue === 'function') {
      return codeMirror.CodeMirror.getValue();
    }
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent) {
      return Array.from(cmContent.querySelectorAll('.cm-line')).map(function (line) { return line.textContent; }).join('\n');
    }
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.from(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 1)] = codeFromCell(cell);
    });
    return code;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch12-lab-feedback-list');
    if (list) list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row ? row.querySelector('span') : null;
      var item = feedback ? feedback[question] : null;
      if (!row || !message || !item) return;
      row.classList.remove('correct', 'incorrect');
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch12-lab-feedback-list');
    if (list) list.hidden = true;
    grader.querySelectorAll('.ch12-lab-feedback-row').forEach(function (row) {
      row.classList.remove('correct', 'incorrect');
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var saved = text ? JSON.parse(text) : null;
        if (!response.ok || !saved || !saved.ok) throw new Error(saved && (saved.message || saved.error) ? (saved.message || saved.error) : 'Submission API is unavailable.');
        return saved;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch12-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var result = document.getElementById('ch12-lab-result');
    var submit = document.getElementById('ch12-lab-submit');
    var actions = grader.querySelector('.ch12-lab-actions');
    var reset = document.getElementById('ch12-lab-reset');

    loadSessionIdentity(identity, actions, submit, result);

    submit.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = 'Submitting code.';
      submit.disabled = true;
      saveAttempt({
        lab_id: labId,
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          if (payload.feedback) applyFeedback(grader, payload.feedback);
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
</script>
