# APIs

APIs let your code communicate with web services, data sources, and cloud tools.

- REST APIs and HTTP methods
- `requests.get()` and response objects
- JSON data as Python dictionaries and lists
- Query parameters and headers
- Reliability patterns for errors, timeouts, pagination, and retries

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | API | An application programming interface, a defined way for software systems to communicate. |
| 2 | Web API | An API accessed over the web, usually with HTTP requests. |
| 3 | HTTP | The protocol commonly used for web requests and responses. |
| 4 | REST | An API design style that uses resources, URLs, and standard HTTP methods. |
| 5 | Endpoint | A URL where a specific API resource or action is available. |
| 6 | HTTP method | A request action such as GET, POST, PUT, PATCH, or DELETE. |
| 7 | Status code | A numeric response code that summarizes whether an HTTP request succeeded or failed. |
| 8 | GET request | An HTTP request used to retrieve data. |
| 9 | POST request | An HTTP request commonly used to send new data to a server. |
| 10 | Query parameter | A key-value option added to a URL to customize a request. |
| 11 | Header | Metadata sent with an HTTP request or response. |
| 12 | API key | A credential used to identify or authorize an API client. |
| 13 | JSON | A text data format commonly used for API request and response bodies. |
| 14 | Response body | The data returned by a server after a request. |
| 15 | `requests` | A popular Python library for making HTTP requests. |
| 16 | Timeout | A limit on how long a client waits for a response. |
| 17 | Pagination | Splitting a large API response into multiple pages or batches. |
| 18 | Exponential backoff | A retry strategy that increases wait time after repeated transient failures. |
| 19 | Retry | Trying a failed request again, often after a short delay. |
| 20 | Response contract | The expected structure and meaning of data returned by an API. |
