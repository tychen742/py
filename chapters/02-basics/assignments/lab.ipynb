```python
### Question 1: Variables and f-strings
# A help desk team tracks one short batch of support tickets.
# Create a variable named remaining_tickets.
# Create a variable named resolution_rate that stores the percent resolved today.
# Print exactly these three lines, using an f-string for each line:
# Team: data services
# Remaining tickets: 13
# Resolution rate: 27.8%
team = "data services"
open_tickets = 18
resolved_today = 5
escalated = 2
### Your code starts here.



### Your code ends here.
```

```python
### Solution
team = "data services"
open_tickets = 18
resolved_today = 5
escalated = 2

remaining_tickets = open_tickets - resolved_today
resolution_rate = resolved_today / open_tickets * 100

print(f"Team: {team}")
print(f"Remaining tickets: {remaining_tickets}")
print(f"Resolution rate: {resolution_rate:.1f}%")
```

```python
### Question 2: Numeric operators
# The variables below describe a small order.
# Calculate subtotal, tax, and total.
# Calculate within_budget as a Boolean value.
# Print exactly these four lines:
# Subtotal: 51.0
# Tax: 4.21
# Total: 55.21
# Within budget: True
notebook_price = 12.50
pen_price = 8.75
sticker_price = 4.25
quantity = 2
tax_rate = 0.0825
budget = 60
### Your code starts here.



### Your code ends here.
```

```python
### Solution
notebook_price = 12.50
pen_price = 8.75
sticker_price = 4.25
quantity = 2
tax_rate = 0.0825
budget = 60

subtotal = (notebook_price + pen_price + sticker_price) * quantity
tax = round(subtotal * tax_rate, 2)
total = round(subtotal + tax, 2)
within_budget = total <= budget

print("Subtotal:", subtotal)
print("Tax:", tax)
print("Total:", total)
print("Within budget:", within_budget)
```

```python
### Question 3: Type conversion and built-in functions
# The values below are strings because they came from a web form.
# Convert each value to an integer.
# Use sum(), len(), max(), and type() to print exactly these four lines:
# Total responses: 46
# Average responses: 11.5
# Largest batch: 15
# Converted first value type: <class 'int'>
day1_text = "12"
day2_text = "9"
day3_text = "15"
day4_text = "10"
### Your code starts here.



### Your code ends here.
```

```python
### Solution
day1_text = "12"
day2_text = "9"
day3_text = "15"
day4_text = "10"

day1 = int(day1_text)
day2 = int(day2_text)
day3 = int(day3_text)
day4 = int(day4_text)
responses = [day1, day2, day3, day4]

print("Total responses:", sum(responses))
print("Average responses:", sum(responses) / len(responses))
print("Largest batch:", max(responses))
print("Converted first value type:", type(day1))
```

```python
### Question 4: Strings and Boolean expressions
# Use the variables below to build a small account check.
# Create greeting by combining "Hello, " with first_name and "!".
# Create first_code as the first character in access_code.
# Create code_prefix as the first two characters in access_code.
# Create eligible as a Boolean expression that is True only when:
# account_active is True, login_count is at least 3, and access_code is not empty.
# Print exactly these four lines:
# Greeting: Hello, Maya!
# First code: A
# Code prefix: A7
# Eligible: True
first_name = "Maya"
access_code = "A7Z9"
account_active = True
login_count = 3
### Your code starts here.



### Your code ends here.
```

```python
### Solution
first_name = "Maya"
access_code = "A7Z9"
account_active = True
login_count = 3

greeting = "Hello, " + first_name + "!"
first_code = access_code[0]
code_prefix = access_code[0:2]
eligible = account_active and login_count >= 3 and bool(access_code)

print("Greeting:", greeting)
print("First code:", first_code)
print("Code prefix:", code_prefix)
print("Eligible:", eligible)
```

```python
### Question 5: Collection types
# Use a list, set, dictionary, and tuple to summarize the fruit data.
# Create unique_fruits by converting fruits to a set.
# Create price_pair as a tuple containing "banana" and its price.
# Print exactly these four lines:
# Fruit count: 4
# Unique fruits: ['apple', 'banana', 'cherry']
# Banana price: 0.5
# Price pair: ('banana', 0.5)
fruits = ["apple", "banana", "apple", "cherry"]
fruit_prices = {"apple": 0.75, "banana": 0.50, "cherry": 1.25}
### Your code starts here.



### Your code ends here.
```

```python
### Solution
fruits = ["apple", "banana", "apple", "cherry"]
fruit_prices = {"apple": 0.75, "banana": 0.50, "cherry": 1.25}

unique_fruits = set(fruits)
price_pair = ("banana", fruit_prices["banana"])

print("Fruit count:", len(fruits))
print("Unique fruits:", sorted(unique_fruits))
print("Banana price:", fruit_prices["banana"])
print("Price pair:", price_pair)
```

<script>
(function () {
  var labId = 'ch02-lab';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';
    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var codeMirror = cell.querySelector('.CodeMirror');
    if (codeMirror && codeMirror.CodeMirror && typeof codeMirror.CodeMirror.getValue === 'function') {
      return codeMirror.CodeMirror.getValue();
    }
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent) {
      return Array.from(cmContent.querySelectorAll('.cm-line')).map(function (line) { return line.textContent; }).join('\n');
    }
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.from(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 1)] = codeFromCell(cell);
    });
    return code;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch02-lab-feedback-list');
    if (list) list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row ? row.querySelector('span') : null;
      var item = feedback ? feedback[question] : null;
      if (!row || !message || !item) return;
      row.classList.remove('correct', 'incorrect');
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch02-lab-feedback-list');
    if (list) list.hidden = true;
    grader.querySelectorAll('.ch02-lab-feedback-row').forEach(function (row) {
      row.classList.remove('correct', 'incorrect');
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var saved = text ? JSON.parse(text) : null;
        if (!response.ok || !saved || !saved.ok) throw new Error(saved && (saved.message || saved.error) ? (saved.message || saved.error) : 'Submission API is unavailable.');
        return saved;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch02-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var result = document.getElementById('ch02-lab-result');
    var submit = document.getElementById('ch02-lab-submit');
    var actions = grader.querySelector('.ch02-lab-actions');
    var reset = document.getElementById('ch02-lab-reset');

    loadSessionIdentity(identity, actions, submit, result);

    submit.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = 'Submitting code.';
      submit.disabled = true;
      saveAttempt({
        lab_id: labId,
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          if (payload.feedback) applyFeedback(grader, payload.feedback);
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
</script>
