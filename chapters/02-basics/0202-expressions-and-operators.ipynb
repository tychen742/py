# Expressions and Operators

An **expression** is code that produces a value. Operators combine values into larger expressions. This section focuses on the operators that appear constantly in Python programs: arithmetic, comparison, Boolean logic, membership, and assignment updates.

```python
subtotal = 18.50
shipping = 4.99
tax_rate = 0.0825

total = (subtotal + shipping) * (1 + tax_rate)
print(round(total, 2))
```

## Arithmetic Operators

Arithmetic operators work with numbers and often appear in data cleaning, measurement conversion, scoring, and simulation.

| Operator | Meaning | Example |
|---|---|---|
| `+` | addition | `3 + 2` |
| `-` | subtraction | `3 - 2` |
| `*` | multiplication | `3 * 2` |
| `/` | true division | `3 / 2` |
| `//` | floor division | `3 // 2` |
| `%` | remainder | `3 % 2` |
| `**` | exponentiation | `3 ** 2` |

```python
minutes = 125
hours = minutes // 60
remaining_minutes = minutes % 60
print(f"{hours} hours and {remaining_minutes} minutes")
```

`//` and `%` are often used together when you need quotient-and-remainder logic.

```python
page_size = 25
record_count = 113
full_pages = record_count // page_size
leftover = record_count % page_size
needs_extra_page = leftover > 0
print(full_pages, leftover, needs_extra_page)
```

## Comparison and Boolean Logic

Comparison operators produce Boolean values: `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | equal value |
| `!=` | not equal value |
| `<`, `<=` | less than, less than or equal |
| `>`, `>=` | greater than, greater than or equal |

```python
score = 87
print(score >= 70)
print(score == 100)
print(score != 0)
```

Boolean operators combine conditions. `and` requires both sides to be true. `or` requires at least one side to be true. `not` reverses a Boolean value.

```python
age = 20
has_id = True
is_member = False

can_enter = age >= 18 and has_id
gets_discount = is_member or age < 21
print(can_enter)
print(gets_discount)
```

Python also treats some non-Boolean values as true or false in conditions. Empty containers, empty strings, numeric zero, and `None` are falsy; most other values are truthy.

```python
values = ["", "Python", 0, 42, [], [1, 2], None]
for value in values:
    print(repr(value), bool(value))
```

## Membership and Identity

Use `in` and `not in` to test whether a value appears in a container.

```python
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)
print("pear" not in fruits)
```

Use `is` for identity checks, especially with `None`. Use `==` for value comparisons.

```python
answer = None
print(answer is None)

numbers = [1, 2, 3]
other_numbers = [1, 2, 3]
print(numbers == other_numbers)
print(numbers is other_numbers)
```

## Assignment Updates and Precedence

Python provides shorthand assignment operators for updating a variable from its current value.

```python
count = 0
count += 1
count += 1
print(count)

price = 100
price *= 0.9
print(price)
```

Operator precedence controls how Python groups expressions. Use parentheses when precedence is not obvious.

| High to low | Operators |
|---|---|
| exponentiation | `**` |
| unary signs | `+x`, `-x` |
| multiplication/division | `*`, `/`, `//`, `%` |
| addition/subtraction | `+`, `-` |
| comparisons | `==`, `!=`, `<`, `<=`, `>`, `>=`, `in`, `is` |
| Boolean not | `not` |
| Boolean and | `and` |
| Boolean or | `or` |

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(not True or True)
print(not (True or True))
```

```python
### Checkpoint: make precedence explicit
# Add parentheses so a reader can immediately see the intended grouping.
result = 12 + 8 / 4 > 10 and 3 * 2 == 6
print(result)
```

## Summary

- Expressions produce values; statements perform actions.

- Comparison and Boolean operators build conditions.
- Membership checks container contents; identity checks object sameness.
- Parentheses are better than relying on a reader to memorize precedence.
