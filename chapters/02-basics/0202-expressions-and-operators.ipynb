### Arithmetic Operators

Arithmetic operators work with numbers and often appear in data cleaning, measurement conversion, scoring, and simulation.

| Operator | Meaning | Example |
|---|---|---|
| `+` | addition | `3 + 2` |
| `-` | subtraction | `3 - 2` |
| `*` | multiplication | `3 * 2` |
| `/` | true division | `3 / 2` |
| `//` | **floor division** | `3 // 2` |
| `%` | **remainder** | `3 % 2` |
| `**` | exponentiation | `3 ** 2` |

Note that:
- `/` true division: always returns a **float**, even 4 / 2 → 2.0
- `//` floor division: returns the **quotient**, rounded down toward negative infinity (not just truncated — matters for negatives)
- `%` modulo: returns the **remainder**
- `**` exponentiation: it's right-associative: 2 ** 3 ** 2 → 2 ** 9 → 512, not 8 ** 2

````python
minutes = 125
hours = minutes / 60
print(f"{hours} hours \t(125/60: true division)")

hours = minutes // 60
print(f"{hours} hours \t\t\t(floor125//60: floor division)")

remaining_minutes = minutes % 60
print(f"{remaining_minutes} minutes \t\t\t(125 % 60: remainder)\n")

print(f"{2**3**2} \t\t\t\t2**3**2: exponentiation")
print(f"{2**(3**2)} \t\t\t\t2**(3**2): exponentiation")
print(f"{(2**3)**2} \t\t\t\t(2**3)**2: exponentiation")
2**3**2        ### exponentiation
````

`//` and `%` are often used together when you need quotient-and-remainder logic.

````python
page_size = 25
record_count = 113
full_pages = record_count // page_size
leftover = record_count % page_size
needs_extra_page = leftover > 0
print(full_pages, leftover, needs_extra_page)
````

### Comparison and Boolean Logic

Comparison operators produce Boolean values: `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | equal value |
| `!=` | not equal value |
| `<`, `<=` | less than, less than or equal |
| `>`, `>=` | greater than, greater than or equal |

````python
score = 87
print(score >= 70)
print(score == 100)
print(score != 0)
````

Boolean operators combine conditions. `and` requires both sides to be true. `or` requires at least one side to be true. `not` reverses a Boolean value.

````python
age = 20
has_id = True
is_member = False

can_enter = age >= 18 and has_id
gets_discount = is_member or age < 21
print(can_enter)
print(gets_discount)
````

Python also treats some non-Boolean values as true or false in conditions: 
- **Falsy**: Empty containers, empty strings, numeric zero, and `None` are falsy;
- **Truthy**: most other values are truthy.

(*`repr()` returns the "official" string representation of an object, one that's meant to be unambiguous and, where possible, could be copy-pasted back into Python to recreate that object.)

````python
values = ["", "Python", 0, 42, [], [1, 2], None]
for value in values:
    print(f"{repr(value):15} {bool(value)}")
````

### Membership and Identity

**Membership** operators `in` and `not in` test whether a value appears in a container (sequence of collection). For examples:


````python
print('a' in 'cat')         # True
print('z' in 'cat')         # False
print('z' not in 'cat')     # True
````

Membership checks work across container types.

````python
# Strings — checks substring
print('ell' in 'hello')         # True
print('z' in 'hello')           # False

# Lists — checks element presence
fruits = ["apple", "banana", "cherry"]
print('apple' in fruits)        # True
print('pear' in fruits)         # False
print()

# Tuples — same as lists
print(3 in (1, 2, 3))           # True

# Dictionaries — checks KEYS by default, not values
d = {'a': 1, 'b': 2}
print('a' in d)                 # True
print(1 in d)                   # False  ← common gotcha
print(1 in d.values())          # True — need .values() explicitly
print()

# Sets
print(3 in {1, 2, 3})           # True
````

Use and `is` and `not is` to test **identity**: whether two names point to the exact same object in memory; not whether their values are equal.

````python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)          # True  — same values
print(a is b)          # False — different objects in memory
````

Use `is` for identity checks, especially with `None`. Use `==` for value comparisons.

````python
answer = None
print(answer is None)

numbers = [1, 2, 3]
other_numbers = [1, 2, 3]
print(numbers == other_numbers)
print(numbers is other_numbers)
````

## Assignment Updates and Precedence

Python provides shorthand assignment operators for updating a variable from its current value.

````python
count = 0
count += 1
count += 1
print(count)

price = 100
price *= 0.9
print(price)
````

Operator precedence controls how Python groups expressions. Use parentheses when precedence is not obvious.

| High to low | Operators |
|---|---|
| exponentiation | `**` |
| unary signs | `+x`, `-x` |
| multiplication/division | `*`, `/`, `//`, `%` |
| addition/subtraction | `+`, `-` |
| comparisons | `==`, `!=`, `<`, `<=`, `>`, `>=`, `in`, `is` |
| Boolean not | `not` |
| Boolean and | `and` |
| Boolean or | `or` |

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(not True or True)
print(not (True or True))
````

````python
### Checkpoint: make precedence explicit
# Add parentheses so a reader can immediately see the intended grouping.
result = 12 + 8 / 4 > 10 and 3 * 2 == 6
print(result)
````

````python
### Solution
result = ((12 + (8 / 4)) > 10) and ((3 * 2) == 6)
print(result)

````

## Summary

- Expressions produce values; statements perform actions.
- Arithmetic operators support calculation and quotient/remainder patterns.
- Comparison and Boolean operators build conditions.
- Membership checks container contents; identity checks object sameness.
- Parentheses are better than relying on a reader to memorize precedence.
