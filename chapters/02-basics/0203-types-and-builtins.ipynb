Types describe what kind of value an object has and what operations make sense for it. Built-in functions are always available and give you a small standard toolkit for inspecting, converting, summarizing, and organizing values.


````python
value = "42"
print(type(value))
print(int(value) + 8)
````

## Core Scalar Types

The most common scalar types are `int`, `float`, `str`, `bool`, and `NoneType`.

| Type | Example | Common use |
|---|---|---|
| `int` | `42` | counts, indexes, whole numbers |
| `float` | `3.14` | measurements, averages, rates |
| `str` | `"Python"` | text |
| `bool` | `True` | conditions and flags |
| `NoneType` | `None` | missing or not-yet-available value |

````python
items = 42
average = 3.14
language = "Python"
finished = False
missing = None

for value in [items, average, language, finished, missing]:
    print(repr(value), type(value).__name__)
````

Strings deserve special attention because they are sequences of characters. You can index and slice strings, but you cannot mutate them in place.

````python
word = "Python"
print(word[0])
print(word[-1])
print(word[0:3])
print(word.lower())
````

## Type Checking and Conversion

Use `type()` when you want to inspect exactly what kind of object you have. Use `isinstance()` when your code needs to accept a family of related types.

````python
print(type(True))
print(isinstance(True, bool))
print(isinstance(True, int))  # bool is a subclass of int in Python
````

Use constructor functions such as `int()`, `float()`, `str()`, `bool()`, `list()`, and `tuple()` to convert values when conversion is valid.

````python
text_count = "15"
count = int(text_count)
rate = float("0.0825")
message = str(count) + " items"

print(count, type(count).__name__)
print(rate, type(rate).__name__)
print(message, type(message).__name__)
````

Python sometimes performs automatic conversion, called **coercion**, in mixed numeric expressions.

````python
result = 3 + 2.5
print(result)
print(type(result).__name__)
````

## Common Built-in Functions

Built-in functions are available without imports. The table below lists a compact working set for early Python programs.

| Function | Purpose |
|---|---|
| `print()` | display output |
| `type()` | inspect an object's type |
| `len()` | count items in a sequence or collection |
| `round()` | round a number |
| `abs()` | absolute value |
| `min()`, `max()` | smallest or largest value |
| `sum()` | add numeric values |
| `sorted()` | return a sorted list |
| `input()` | read text from a user in interactive programs |

````python
scores = [88, 92, 75, 100, 83]
print(len(scores))
print(min(scores), max(scores))
print(sum(scores) / len(scores))
print(sorted(scores))
````

Some functions return values you can store and reuse. `print()` displays output but returns `None`.

````python
rounded_average = round(sum(scores) / len(scores), 1)
print(rounded_average)

return_value = print("display only")
print(return_value)
````

## Collection Type Preview

Collections hold multiple values. This chapter only introduces the main collection types; later chapters develop them in depth.

| Type | Literal form | Main idea | Later chapter |
|---|---|---|---|
| `list` | `[1, 2, 3]` | ordered, mutable sequence | Chapter 6 |
| `tuple` | `(1, 2, 3)` | ordered, immutable sequence | Chapter 6 |
| `dict` | `{"name": "Ada"}` | key-value mapping | Chapter 7 |
| `set` | `{1, 2, 3}` | unordered unique values | Chapter 6 |

````python
fruits = ["apple", "banana", "apple", "cherry"]
unique_fruits = set(fruits)
student = {"name": "Maya", "score": 92}
point = (3, 5)

print(fruits)
print(unique_fruits)
print(student["name"])
print(point[0])
````

Choose a collection based on the job: list for ordered data you may change, tuple for fixed records, dictionary for named fields or lookup by key, and set for uniqueness or membership.

## Type Hints

Type hints document the types a programmer expects. Python does not enforce these hints at runtime by default, but editors and checkers can use them to catch mistakes earlier.

````python
def format_score(name: str, score: float) -> str:
    return f"{name}: {score:.1f}"

print(format_score("Maya", 92))
````

````python
### Checkpoint: add type hints
# Add type hints to the function below.
def total_price(quantity, price):
    return quantity * price

print(total_price(3, 4.99))
````

````python
### Solution
def total_price(quantity: int, price: float) -> float:
    return quantity * price

print(total_price(3, 4.99))

````
