<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Variable | A name that refers to a value or object in a program. |
| 2 | Assignment statement | A statement that binds a variable name to a value, such as `score = 95`. |
| 3 | Naming Rules & Conventions | Shared rules and recommended styles for variable names and other identifiers. |
| 4 | Keyword | A reserved word with special meaning in Python that cannot be used as an ordinary variable name. |
| 5 | Soft keyword | A word that acts like a keyword only in certain Python contexts but can still be used as a name elsewhere. |
| 6 | Object | A Python value with an identity, a type, and data. |
| 7 | Identity | The unique object reference Python uses to tell whether two names point to the same object. |
| 8 | Alias | A second name that refers to the same object as another name. |
| 9 | Type | The category of a value, such as `int`, `float`, `str`, `bool`, `list`, or `dict`. |
| 10 | Operator | A symbol or word that performs an operation, such as `+`, `==`, `and`, `or`, or `in`. |
| 11 | Operand | A value that an operator works on. |
| 12 | Expression | A piece of code that evaluates to a value. |
| 13 | Built-in function | A function available in Python without an import, such as `len`, `print`, `type`, or `round`. |
| 14 | Argument | A value passed into a function call. |
| 15 | Return value | The value produced by a function call. |
| 16 | Type conversion | The process of changing a value from one type to another. |
| 17 | Type hint | An annotation that records the expected type of a variable, parameter, or return value. |
| 18 | Mutable | Describes an object whose contents can be changed after creation. |
| 19 | Immutable | Describes an object whose contents cannot be changed after creation. |
| 20 | List | An ordered, mutable collection of values. |
| 21 | Tuple | An ordered, immutable collection of values. |
| 22 | Dictionary | A mutable collection that stores key-value pairs. |
| 23 | Set | A mutable collection that stores unique values without a fixed order. |
| 24 | Boolean | A type with two possible values: `True` and `False`. |
| 25 | Truthy | Describes a value that behaves like `True` in a Boolean context. |
| 26 | Falsy | Describes a value that behaves like `False` in a Boolean context. |
