## Variables

Programs need a way to remember values. In Python, a **variable** is a named location in computer memory that refers to an **object**. The object stores the **value**; the name lets your program use that value again. Think of a variable as a name/label that is associated with some data. Variables are fundamental to programming because they allow us to store, retrieve, and manipulate data throughout our program. 

Unlike some other programming languages, Python doesn't require you to **declare** the **data type** of a variable before using it. The type is automatically determined based on the value you assign (this is called **dynamic typing**).



### Creating Variables

In Python, you create a variable by using an **assignment statement** with the assignment operator `=`. Read the assignment statement syntax below as "**assign the value on the right to the name on the left.**" Python evaluates the right side first, then stores the reference in the name.

```python
variable_name = value (an expression)
```

An assignment statement has three parts: 

1. the **name** of the variable on the left,
2. the **assignment operator**, **`=`**, and
3. an **expression** (the value) on the right.


Note that:
1. A **variable** is a name that refers to a value.
2. When you run an assignment **_statement_**, there is **no output**. Python creates the variable and gives it a value, but the assignment statement has no visible effect.
3. A literal value is also an expression.
4. After creating a variable, you can use it as an expression. 
5. Variables can be **reassign**ed.
6. You can assign **multiple** values/expressions to multiple variables in one line.

For example, you can create variables as follows.

````python
x = 1
age = 20
name = "Alice"
pi = 3.14
is_active = True
````


You can assign **multiple** values/expressions to multiple variables in one line. For example:
- assign multiple variables the same value
- assign multiple variables different values

````python
x = y = z = 0           # x, y, and z are all 0
a, b, c = 10, 20, 30    # assign different values to variables

print(f"x = {x}, y = {y}, z = {z}")
print(f"a = {a}, b = {b}, c = {c}")
````

Or, you can swap variable values by switching the positions of variables.

````python
m, n = 5, 10            # m is 5 and n is 10
print(f"Before swap:\t m = {m}, n = {n}")

m, n = n, m             # Swap values using tuple unpacking ==> sorting 
print(f"After swap:\t m = {m}, n = {n}")
````

A name can be rebound to a different object. **Reassignment** does not edit the old object; it makes the name refer to another object.

````python
x = 10         # x is 10
x = 20.5       # now x is 20.5
x = "hello"    # now x is a string (type can change!)
print(x)
````

````python
label = "draft"
print(label)

label = "final"
print(label)

print("label was", label, "and is now", label)
print(f"label was {label} and is now {label}")  # f-string
````

### Naming Conventions

General Python variable naming follows the convention from the official [PEP 8 style guidelines](https://peps.python.org/pep-0008/#naming-conventions), although conventions such as Google Python:

1. Length: Names can be as long as you like.
2. Containing: Names can contain alphanumeric characters (letters and numbers) and underscores. Special characters (e.g., @, #, $) are not allowed.
3. Begin with: Can begin with an **underscore** or a **letter**, but can't begin with a number.
4. Letter Case:
   1. Names are case-sensitive. `age`, `Age`, and `AGE` are three distinct variables.
   2. It is conventional to use only **lowercase** letters for variable names, although it is legal to use uppercase letters.
   3. Use **`snake_case`** for variable names: All lowercase with **underscores** between words. Example: total_price, student_name, is_active. (camelCase is for C#/Java/JavaScript variable names.)
   4. For class names, use **PascalCase**, or `upper camel case`, convention; e.g., `class StudentRecord`. 
7. Avoid **keywords** and **built-in function** names for naming variables.
8. Constants: For constants, use ALL_CAPS with underscores; e.g., MAX_SIZE = 100, PI = 3.14159.

The only punctuation that can appear in a variable name is the underscore character, `_`. It is often used in names with multiple words, such as `your_name` or  `airspeed_of_unladen_swallow`.

If you give a variable an illegal name, you get a syntax error. For example, the name `million!` is illegal because it contains punctuation.

Python programmers follow these **conventions** (recommended but not required):

| Convention | Use Case | Example |
|------------|----------|---------|
| **snake_case** | Variable and function names | `student_name`, `total_price` |
| UPPER_CASE | Constants | `MAX_SIZE`, `PI` |
| PascalCase (not camelCase) | Class names | `StudentRecord`, `BankAccount` |
| **Descriptive** names | Make code readable | `count` instead of `c` |

Good naming makes code easier to scan and therefore is great for **documentation**. In Python, variable names usually use **`snake_case`**: lowercase words separated by underscores.

| Pattern | Use |

| `student_count` | regular variable |

Avoid names that shadow common built-ins, such as `list`, `sum`, `str`, or `type`.

## Python Objects

In Python, [everything is an **object**](https://www.pythonmorsels.com/everything-is-an-object/), including classes, instances of classes, modules, and functions. In short, anything that you can point a **variable** to is an **object**. Variables in Python do not **save** values as *buckets* containing things; they're pointers pointing to objects.



In Python, "everything is an **object**", and every object has a value, a type, and an identity: 

- **identity** (unique ID) (`id()`): Each object has a unique identifier that remains constant for the object's lifetime. This is the object's memory address and can be retrieved using the built-in id() function.
- **value**: This represents the data that the object stores. 
- **type** (data type): An object's type defines what kind of data it represents and what operations can be performed on it. The type of an object can be determined using the built-in type() function, and every object is an instance of some class. 


You can use `type()` to inspect the kind of data the object refers to, and use `id()` to find the identification the object has.

### Data Type

The values that are associated with the object can be checked by using the`type()` function.

````python
### check data type

items = 12
price = 4.99
name = "notebook"
in_stock = True

print(type(items))
print(type(price))
print(type(name))
print(type(in_stock))
````

### Object ID
In Python, every object has a unique identity, which can be obtained using the built-in `id()` function. An identity is a unique integer that remains constant for an object throughout its lifetime. This is useful for understanding how Python manages objects in memory and for distinguishing between objects that have the same value. 

You can use the `id()` function when you need to reason about identity. In the examples below, you see that different variables have different IDs.

````python
### different variables (objects) have different IDs

num1 = 42
num2 = 3.14
greeting = "hello"
fruits = ["apple", "banana", "cherry"]

print(id(num1))
print(id(num2))
print(id(greeting))
print(id(fruits))
````

### Mutability

Some objects can be changed after the variables are created, and some can not:
- **Mutable**: Data collection types such as `list` and `dict` (dictionary) are mutable. After creating variables with such data, you may change (update) the data. 
- **Immutable**: Literal data types such as integers, floats, strings, tuples, booleans, and `None`, and other collection types are immutable. You can not change (update) the value after the variables are created. In order to change the object value to the variable name, you have to perform a new assignment (reassignment). 

| Action | What actually happens | Identity (`id()`) |
|---|---|---|
| **Reassignment** (`x = new_value`) | `x` is repointed to a *different* object entirely | Changes |
| **Mutation** (`x.append(...)`, `x[0] = ...`) | The *same* object's contents change in place | Stays the same |

````python
### mutation: different value, same ID

x = [1, 2, 3]

print(x, id(x))

x.append(4)     ### mutation; data (object) updated

print(x, id(x))
````

````python
### reassignment: different value, different ID

num = 5
print(num, id(num))

num = 5 + 1      ### reassignment; new object created
print(num, id(num))
````

**A note on mutation

In principle, each variable assignment operation creates a unique ID. 

Note that, for **immutable** types, Python may bind the same value to different variables with the same type to save memory. Therefore, different variables with the same value **may** have the same ID when the data is small in size. 

In the example below, small integer literals are *interned* — reused as the same object — as a memory optimization. This is a CPython implementation detail, not something your code should ever rely on. 

````python
### small int, same IDs

num1=100
num2=100
print(id(num1))
print(id(num2))
````

````python
### large int, different IDs

num1=1000
num2=1000
print(id(num1))
print(id(num2))
````

````python
### None, same ID

a = None
b = None
print(id(a))
print(id(b))
````

### Aliasing
Aliasing happens when two or more variable names refer to the **same object** in memory, rather than each having its own independent copy. Because they point to the same object, changing the object through one name is visible through the other name too. 

When two names refer to the same **mutable object**, changing the object through one name affects what you see through the other name. This is called **aliasing**, meaning different variables with the same value will have not only the same values but also the same IDs.



Let's start with creating two `list` variables by using two assignment statements and observe their values and IDs. You can see that they have the same values but with different IDs.

````python
a = [10, 20, 30]
b = [10, 20, 30]

print(a)
print(b)

print(id(a))
print(id(b))
````

Now we'll check their value equality and identity by using the keyword `is` and the operator `==`. You see that they have the same value, but they are different objects; which means the value binds to list `b` is a different copy from list `a`. They just happen to be the same value. 

````python
print(a == b)             ### True; same value
print(a is b)             ### False; different ID == different objects
````

Since variables are just names, two variables can point to the same object — and if that object is mutable, changing it through one name is visible through the other.

In the example below, you can see that both list `a` and `b` have the same value and the same ID. 

````python
### a and b have the same value and the same ID
a = [10, 20, 30]
b = a     ### b is now an ALIAS for a — same object, not a copy

print(a)
print(b)

print(id(a))
print(id(b))
````

Since they have the same ID, both variable names point to the same object. The data variable `a` points to is the same copy `b` points to -- there's only one copy (same object).

#### Sameness

Here, you use the `==` operator to check value equality and the `is` keyword to check object identity. 

- **Equality** Operators: **`==`** and **`!=`**. These check if two values are equal (have the same content).
- **Identity** Operator: **`is`**. This checks if two variables point to the same object in memory (aliasing!).

As a summary of the equity and identiity operators:

| Operator | Purpose | Example | What it checks |
|----------|---------|---------|----------------|
| `==` | Equality | `[1,2] == [1,2]` | Do they have the same content? |
| `!=` | Inequality | `[1,2] != [3,4]` | Do they have different content? |
| `is` | Identity | `x is y` | Do they point to the same object? |
| `is not` | Non-identity | `x is not y` | Do they point to different objects? |

Or, simply:

| Operator | Checks   | Meaning     |
| -------- | -------- | ----------- |
| `is`     | Identity | Same object |
| `==`     | Equality | Same value  |


````python
print(a == b)             ### True; same value
print(a is b)             ### True; same ID == same object
````

The surprising side effect is that if you update the value of one variable, the other variable will reflect the change as well, since they are the same object.

````python
a = [10, 20, 30]
b = a                     ### b is an "alias" of a

print(a)                  ### [10, 20, 30]
print(b)                  ### [10, 20, 30]

print()

print("a == b: ", a == b)     ### '==' for equality (same value) test
print("a is b: ", a is b)     ### '==' for equality (same value) test
print("id(a) == id(b) ", id(a) == id(b))
print(a, id(a))         ### a is changed as well because a and b are the same object
print(b, id(b))

print()

print("b.append(40)")
b.append(40)     ### change b

print()

print(a, id(a))         ### a is changed as well because a and b are the same object
print(b, id(b))
````

Another example of aliasing.

````python
nums = [1, 2, 3]
print(f"nums is {nums} and has id {id(nums)}")

alias = nums    ### Create an alias for nums
print(f"alias is {alias} and has id {id(alias)}")

print("\nmodifying the list using the alias...")
alias.append(4) ### Modify the list using the alias
print(f"\nalias is now {alias} and has id {id(alias)}")

print(f"nums is now {nums} too, and has id {id(nums)}")
````

This side effect does not happen with literal data types such as strings, which are **immutable**. 

For `list` and `dict` collection types, to avoid the side effect of **mutable** aliasing, you may:

- Use **reassignment** with the alias variable. Reassignment moves one name to a new object; it does not change the original object.
- Use the .copy() method to create an independent copy of the original object.

````python
nums = [1, 2, 3]
alias = nums

alias = [1, 2, 3, 4]    #### reassignment to alias
print(nums)
print(alias)
````
