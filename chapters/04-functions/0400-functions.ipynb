# Functions

Functions let you name a reusable block of code and call it whenever you need that behavior.

- Functions reduce repetition by giving a name to a reusable process
- Parameters let a function work with different inputs
- Return values let a function produce a result for later code to use
- Scope controls where names are available
- Recursion solves a problem by reducing it to smaller versions of itself

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Function | A named, reusable block of code that performs a task. |
| 2 | Function definition | Code that creates a function using `def`, a name, parameters, and a body. |
| 3 | Function call | An expression that runs a function by using its name followed by parentheses. |
| 4 | Function body | The indented block of statements that runs when a function is called. |
| 5 | Parameter | A variable name listed in a function definition. |
| 6 | Argument | A value passed into a function call. |
| 7 | Default parameter | A parameter that has a fallback value when the caller does not provide an argument. |
| 8 | `*args` | Syntax that collects extra positional arguments into a tuple. |
| 9 | `**kwargs` | Syntax that collects extra keyword arguments into a dictionary. |
| 10 | Return value | The result a function sends back to the caller. |
| 11 | Return statement | A statement that exits a function and optionally sends a value back. |
| 12 | Type annotation | A note in code that records the expected type of a parameter or return value. |
| 13 | Scope | The region of a program where a name can be used. |
| 14 | Local variable | A variable created inside a function and available only within that function call. |
| 15 | Function composition | Building a result by using one function's output as another function's input. |
| 16 | Docstring | A string that documents what a function does. |
| 17 | Lambda function | A small anonymous function written as a single expression. |
| 18 | Recursion | A pattern where a function calls itself. |
| 19 | Stack diagram | A diagram that shows active function calls and their local variables. |
| 20 | Infinite recursion | Recursion that never reaches a stopping condition. |
