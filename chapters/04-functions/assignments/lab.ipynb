```python
### Question 1: Default parameters
# Write format_invoice(amount, currency="USD") that returns a string like
# "120.00 USD", formatting amount to 2 decimal places.
# Call it once without currency and once passing currency="EUR".
# Print exactly these two lines:
# 120.00 USD
# 85.50 EUR
### Your code starts here.



### Your code ends here.
```

```python
### Solution
def format_invoice(amount, currency="USD"):
    return f"{amount:.2f} {currency}"

invoice1 = format_invoice(120)
invoice2 = format_invoice(85.5, "EUR")

print(invoice1)
print(invoice2)
```

```python
### Question 2: Flexible arguments with *args and **kwargs
# Write trip_summary(destination, *stops, **details) that returns a dictionary
# with keys "destination", "stops" (a tuple), and "details" (a dict).
# Call it with destination="Denver", stops "Omaha" and "Boulder",
# and keyword details days=5, budget=900. Print the returned dictionary.
# Print exactly this line:
# {'destination': 'Denver', 'stops': ('Omaha', 'Boulder'), 'details': {'days': 5, 'budget': 900}}
### Your code starts here.



### Your code ends here.
```

```python
### Solution
def trip_summary(destination, *stops, **details):
    return {
        "destination": destination,
        "stops": stops,
        "details": details,
    }

summary = trip_summary("Denver", "Omaha", "Boulder", days=5, budget=900)
print(summary)
```

```python
### Question 3: Function composition for a BMI report
# Write three functions:
# 1. compute_bmi(weight_kg, height_m): returns weight_kg / height_m ** 2
# 2. bmi_category(bmi): returns "underweight" if bmi < 18.5,
#    "healthy" if bmi < 25, otherwise "above healthy range"
# 3. bmi_report(weight_kg, height_m): calls both and returns
#    "BMI is X.X (category)", with bmi rounded to 1 decimal place
# Call bmi_report(70, 1.75) and bmi_report(90, 1.70), printing each result.
# Print exactly these two lines:
# BMI is 22.9 (healthy)
# BMI is 31.1 (above healthy range)
### Your code starts here.



### Your code ends here.
```

```python
### Solution
def compute_bmi(weight_kg, height_m):
    return weight_kg / height_m ** 2

def bmi_category(bmi):
    if bmi < 18.5:
        return "underweight"
    if bmi < 25:
        return "healthy"
    return "above healthy range"

def bmi_report(weight_kg, height_m):
    bmi = compute_bmi(weight_kg, height_m)
    category = bmi_category(bmi)
    return f"BMI is {bmi:.1f} ({category})"

print(bmi_report(70, 1.75))
print(bmi_report(90, 1.70))
```

```python
### Question 4: Pure function refactor
# apply_discount_in_place(prices, percent) below mutates its list argument.
# Write a pure version, apply_discount(prices, percent), that returns a NEW
# list of prices reduced by percent (each rounded to 2 decimals) without
# changing the original list.
# Call apply_discount(cart_prices, 10) and store the result in discounted.
# Print cart_prices, then discounted.
# Print exactly these two lines:
# [40.0, 15.0, 60.0]
# [36.0, 13.5, 54.0]
def apply_discount_in_place(prices, percent):
    for i in range(len(prices)):
        prices[i] = round(prices[i] * (1 - percent / 100), 2)

cart_prices = [40.0, 15.0, 60.0]
### Your code starts here.



### Your code ends here.
```

```python
### Solution
def apply_discount_in_place(prices, percent):
    for i in range(len(prices)):
        prices[i] = round(prices[i] * (1 - percent / 100), 2)

def apply_discount(prices, percent):
    return [round(price * (1 - percent / 100), 2) for price in prices]

cart_prices = [40.0, 15.0, 60.0]
discounted = apply_discount(cart_prices, 10)

print(cart_prices)
print(discounted)
```

```python
### Question 5: Recursion over a list
# Write list_product(nums), which returns the product of all numbers in nums,
# using recursion. The product of an empty list is 1; otherwise multiply the
# first item by the recursive product of the rest of the list.
# Print exactly these two lines:
# 24
# 5
### Your code starts here.



### Your code ends here.
```

```python
### Solution
def list_product(nums):
    if not nums:
        return 1
    return nums[0] * list_product(nums[1:])

print(list_product([2, 3, 4]))
print(list_product([5]))
```

<script>
(function () {
  var labId = 'ch04-lab';

  function getParams() {
    var params = new URLSearchParams(window.location.search);
    return {
      canvas_course_id: params.get('canvas_course_id') || params.get('course_id') || '',
      canvas_assignment_id: params.get('canvas_assignment_id') || params.get('assignment_id') || '',
      canvas_user_id: params.get('canvas_user_id') || params.get('user_id') || ''
    };
  }

  function loadSessionIdentity(identity, actions, submit, result) {
    submit.disabled = true;
    actions.classList.add('is-locked');
    result.textContent = 'Sign in from the account menu to submit this assignment.';
    return fetch('/api/v1/session.php', { credentials: 'same-origin' })
      .then(function (response) { return response.ok ? response.json() : null; })
      .then(function (payload) {
        if (!payload || !payload.authenticated || !payload.identity) return identity;
        identity.canvas_user_id = payload.identity.canvas_user_id || identity.canvas_user_id || '';
        identity.student_identifier = payload.identity.student_identifier || '';
        submit.disabled = false;
        actions.classList.remove('is-locked');
        result.textContent = payload.identity.display_name ? 'Signed in as ' + payload.identity.display_name + '.' : 'Signed in.';
        return identity;
      })
      .catch(function () { return identity; });
  }

  function codeFromCell(cell) {
    var codeMirror = cell.querySelector('.CodeMirror');
    if (codeMirror && codeMirror.CodeMirror && typeof codeMirror.CodeMirror.getValue === 'function') {
      return codeMirror.CodeMirror.getValue();
    }
    var cmContent = cell.querySelector('.cm-content');
    if (cmContent) {
      return Array.from(cmContent.querySelectorAll('.cm-line')).map(function (line) { return line.textContent; }).join('\n');
    }
    var textarea = cell.querySelector('textarea');
    if (textarea) return textarea.value;
    var pre = cell.querySelector('.cell_input pre');
    return pre ? pre.textContent : '';
  }

  function collectCodeCells() {
    var cells = Array.from(document.querySelectorAll('div.cell.tag_thebe-interactive')).slice(0, 5);
    var code = {};
    cells.forEach(function (cell, index) {
      code['q' + (index + 1)] = codeFromCell(cell);
    });
    return code;
  }

  function applyFeedback(grader, feedback) {
    var list = grader.querySelector('.ch04-lab-feedback-list');
    if (list) list.hidden = false;
    ['q1', 'q2', 'q3', 'q4', 'q5'].forEach(function (question) {
      var row = grader.querySelector('[data-question=' + question + ']');
      var message = row ? row.querySelector('span') : null;
      var item = feedback ? feedback[question] : null;
      if (!row || !message || !item) return;
      row.classList.remove('correct', 'incorrect');
      message.textContent = item.message + ' (' + item.score + ' / ' + item.max_score + ')';
      row.classList.add(item.correct ? 'correct' : 'incorrect');
    });
  }

  function clearFeedback(grader) {
    var list = grader.querySelector('.ch04-lab-feedback-list');
    if (list) list.hidden = true;
    grader.querySelectorAll('.ch04-lab-feedback-row').forEach(function (row) {
      row.classList.remove('correct', 'incorrect');
      var message = row.querySelector('span');
      if (message) message.textContent = '';
    });
  }

  function saveAttempt(payload) {
    return fetch('/api/v1/lab-attempts.php', {
      method: 'POST',
      headers: { 'Content-Type': 'application/json' },
      credentials: 'same-origin',
      body: JSON.stringify(payload)
    }).then(function (response) {
      return response.text().then(function (text) {
        var saved = text ? JSON.parse(text) : null;
        if (!response.ok || !saved || !saved.ok) throw new Error(saved && (saved.message || saved.error) ? (saved.message || saved.error) : 'Submission API is unavailable.');
        return saved;
      });
    });
  }

  function initLabGrader() {
    var grader = document.getElementById('ch04-lab-grader');
    if (!grader || grader.dataset.ready === 'true') return;
    grader.dataset.ready = 'true';

    var identity = getParams();
    var result = document.getElementById('ch04-lab-result');
    var submit = document.getElementById('ch04-lab-submit');
    var actions = grader.querySelector('.ch04-lab-actions');
    var reset = document.getElementById('ch04-lab-reset');

    loadSessionIdentity(identity, actions, submit, result);

    submit.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = 'Submitting code.';
      submit.disabled = true;
      saveAttempt({
        lab_id: labId,
        code: collectCodeCells(),
        canvas_course_id: identity.canvas_course_id,
        canvas_assignment_id: identity.canvas_assignment_id,
        canvas_user_id: identity.canvas_user_id
      })
        .then(function (payload) {
          if (payload.feedback) applyFeedback(grader, payload.feedback);
          var best = (payload.best_score === null || typeof payload.best_score === 'undefined') ? payload.score : payload.best_score;
          var attempt = (payload.attempt_count === null || typeof payload.attempt_count === 'undefined') ? '' : ' Attempt ' + payload.attempt_count + '.';
          result.textContent = 'Submission saved.' + attempt + ' Score: ' + payload.score + ' / ' + payload.max_score + '. Best score: ' + best + ' / ' + payload.max_score + '. Canvas sync: ' + payload.canvas_sync_status + '.';
        })
        .catch(function (error) {
          result.textContent = 'Could not save this attempt: ' + error.message;
        })
        .finally(function () {
          submit.disabled = false;
        });
    });

    reset.addEventListener('click', function () {
      clearFeedback(grader);
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initLabGrader);
  } else {
    initLabGrader();
  }
})();
</script>
