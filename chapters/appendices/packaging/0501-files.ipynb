def download(url):
    filename = basename(url)

        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

## The `with` Statement

Before writing to or reading from a file, Python needs to *open* it — and when you're done, the file must be *closed* to flush data and release system resources. Forgetting to close a file can cause data loss or resource leaks.

The `with` statement handles this automatically:

```python
with open('data.txt') as f:
    content = f.read()
# file is closed here automatically, even if an error occurred inside the block
```

This is equivalent to:

```python
f = open('data.txt')
try:
    content = f.read()
finally:
    f.close()          # runs even if an exception is raised
```

Any object that supports the `with` statement is called a **context manager**. `open()` returns one. The `with` block guarantees that setup (opening) and teardown (closing) happen in the right order, even if something goes wrong.

:::{note}
You will see `with` used throughout this course with files, database connections, network sockets, and more. Ch13 explains how to build your own context managers.
:::


````python
text = """Hello, this is the first line of a new text file.
This is the second line."""

with open('../../../data/text_file.txt', 'w') as f:
    f.write(text)

with open('../../../data/text_file.txt') as f:
    file_content = f.read()       ### file object
    print(file_content)
````

## `Path()`

The `Path()` function is from the pathlib module — the modern way to work with file paths in Python. To use it, you 

- import `Path()` from Python's pathlib module
- build the path pointing to the directory or to the file.

````python
from pathlib import Path

data = project_root / 'data'                  ### path of data folder
words_path = project_root / 'data' / 'words.txt'   ### path of words.txt file
````

After you have the path object, you can do many things with it.

````python
# Building paths (works on Windows and Mac/Linux)
f = data / "words.txt"          ### use data folder and file name to build path

# File info
print(f.name)          # "words.txt"
print(f.stem)          # "words"
print(f.suffix)        # ".txt"
print(data.parent)     # parent directory of data/

# Checking existence
print(f.exists())
print(f.is_file())
print(f.is_dir())

# Reading and writing
### uncomment the following line to read the file
### the following line will throw an error if the file does not exist
# print(f.read_text())
tmp = Path('tmp_example.txt')
tmp.write_text("hello")         # write to a temp file, not words.txt
tmp.unlink()                    # clean up

# Current directory and home
print(Path.cwd())      # current working directory
print(Path.home())     # home directory (~)

# Listing files
print(list(data.iterdir()))            # all files in folder
print(list(data.glob("*.csv")))        # all CSVs
print(list(data.rglob("*.txt")))       # recursive
````
