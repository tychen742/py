# Python installation


```{contents}
:local:
:depth: 2
```

Unlike Unix-like systems (macOS and Linux), Windows does not ship with a default Python. It is, however, very common for your computer, Windows or not, to have multiple Python releases installed already. That could have happened because Python adoption has been prevalent and, overtime, the software packages (e.g., Visual Studio IDE, Anaconda, etc.) you use may have bundled Python as a dependency when you install them.


:::{admonition} Terminal & Shell

This section requires fundamental knowledge about command line interface (CLI), including how to launch the terminal application and how to operate the operating system shell by issuing commands in the system shell.

- **Terminal**: A terminal is the program you open to access the **shell**. Common terminal programs include PowerShell/Command Prompt on Windows, Terminal on macOS, and Terminal on Linux.
- **Shell**: The shell is what actually interprets and runs your commands, acting as the interface between you and the operating system. Common shells include `PowerShell`/Command Prompt (Windows), `bash` (macOS and Linux), and zsh (macOS).
- **REPL**: A REPL (Read-Eval-Print Loop) is an interactive environment for a specific programming language, not the OS. You type a line of code, it runs immediately, and the result prints right away. Examples include the **Python shell** (`>>>`at the command line), `csharprepl` for C#, and Node.js's built-in REPL. Jupyter Notebook is also built on a REPL: each notebook cell is a REPL to make each notebook a persistent, non-linear, editable document where cells can be run out of order and outputs can include plots and rich media, not just text.
:::


## Check Existing Python


Since Python is widely used by other tools and packages, you more than likely have some versions of Python automatically installed in your system when installing other packages. To check which version(s) of Python are already installed on your computer, start the terminal application. 

:::{warning}
If you do not have Python installed and you type `python` in Windows Search bar or `python` in PowerShell then enter, Windows will show you a Microsoft Store suggestion for Python installation. **DO NOT install Python from Microsoft Store.** It usually leads to issues with Python execution permission, `pip` installer, and lack of `py.exe` launcher. Go to next section: [Install Python](#install-python) to learn how to download a standalone Python installer from [python.org](python.org) and install Python.
:::

### Start Terminal
Start the terminal application in your computer to use the CLI.

::::{tab-set}
:::{tab-item} Windows

Launch the terminal application (`Windows PowerShell`) so you can type commands at the system shell prompt (`>` with a flashing vertical bar `|`). 

```powershell
Windows PowerShell
Copyright (C) Microsoft Corporation. All rights reserved.

Loading personal and system profiles took 1119ms.
PS C:\Users\[user]> |
```
:::

::::{tab-item} macOS
From Spotlight Search, look for the terminal application (`Terminal.app`) and return to launch it so you can type commands at the system shell prompt (`$` or `%` followed by a block):
```bash
Last login: Fri Aug 28 12:18:57 on ttys000

[user]@[computer]:~$ █
```
::::
::::

### Check Default Python

Now you are at the command line, issue the commands to check the default Python release. 

::::{tab-set}
:::{tab-item} Windows

To check the default Python version, issue `python --version` and enter. If a Python is installed, you will see the returned Python version information.
```powershell
PS C:\Users\[user]> python --version
Python 3.13.7
```

If no Python is installed, you may see the following error message. Go to next section: [Install Python](#install-python) to learn how to download a standalone Python installer from [python.org](python.org) and install Python.
```powershell
PS C:\Users\[user]>
Python was not found; run without arguments to install from the Microsoft Store, or disable this shortcut from Settings > Apps > Advanced app settings > App execution aliases.

PS C:\Users\[user]>
```
:::

:::{tab-item} macOS
With macOS, you already have a version of Python shipped with the OS. You may issue the following commands in the system shell to see all the versions of Python you have in the current system. Note that, in most cases, you may need to use `python3` instead of `python`. 

To find the default Python version when you enter `python3` (or `python`) when you enter it at the command line, pass the `--version` option to Python.
```bash
[user]@[host]:~$ python3 --version
Python 3.12.2
```
:::
::::

### Check Python Installations

In addition to the default installation, you can check how many Python releases are installed in your computer. 

::::{tab-set}
:::{tab-item} Windows
You can use the command-finder (`where`.exe) and the **Python Launcher** for Windows (`py`.exe) to see the current Python installation(s) in Windows PowerShell. 

The `where.exe` command finder shows all Python installations and their installation paths:

```powershell
PS C:\Users\[user]> where.exe python
PS C:\Users\[user]\workspace\py> where.exe python
C:\Users\[user]\AppData\Local\Programs\Python\Python312\python.exe
C:\Users\[user]\AppData\Local\Programs\Python\Python313\python.exe
C:\Users\[user]\AppData\Local\Programs\Python\Python310\python.exe
C:\Users\[user]\AppData\Local\Microsoft\WindowsApps\python.exe
PS C:\Users\[user]\workspace\py>
```

The Python Launcher shows the version information and the *launcher default* Python version (the entry with `*`). You can use `py -0` or `py --list` for this.
```powershell
PS C:\Users\[user]> py -0
 -V:3.13          Python 3.13 (64-bit)
 -V:3.13-arm64    Python 3.13 (Store)
 -V:3.12          Python 3.12 (64-bit)
 -V:3.10 *        Python 3.10 (64-bit)
PS C:\Users\[user]>
```
:::

:::{tab-item} macOS

To check all the installed python versions and their installation paths, use the `which` command (note that this gives you the installations records in the `PATH` environment variable):
```bash
[user]@[host]:~$ which -a python python3
/opt/homebrew/bin/python3
/usr/local/bin/python3
/usr/bin/python3
/opt/homebrew/bin/python3
[user]@[host]:~$
```

To see the version information for the installations above, use the following code and enter.

```bash
[user]@[host]:~$
for p in $(which -a python python3); do
    echo "$p -> $($p --version 2>&1)"
done
[user]@[host]:~$
```

You will see the output with version information as below.
```bash
[user]@[host]:~$
for p in $(which -a python python3); do
    echo "$p -> $($p --version 2>&1)"
done
/opt/homebrew/bin/python3 -> Python 3.14.5
/usr/local/bin/python3 -> Python 3.12.10
/usr/bin/python3 -> Python 3.9.6
/opt/homebrew/bin/python3 -> Python 3.14.5
[user]@[host]:~$
```

To see the Python installations through Homebrew (recommended), issue the commands: 
```bash
[user]@[host]:~$ brew list | grep python@
python@3.12
python@3.13
python@3.14
```

Actually, all your Homebrew Python installations are in `/opt/homebrew/Cellar/` with symbolic links in `/opt/homebrew/opt/`. You can `cd` into those folders and `ls -l python*` to see all of them listed. 
:::

## Install Python

If there is not current Python installation in your computer, or if you would like to install a more current version of Python, go to [python.org](https://python.org) and download the installer for the specific version, then install it. 

```bash
$ python3
Python 3.12.10 (main, May 22 2025, 21:57:53) [Clang 14.0.0 (clang-1400.0.29.202)] on darwin
Type "help", "copyright", "credits" or "license" for more information.
>>>
```

If you are a Windows user, the situation is somewhat more complex. When you search for Python from the Windows search bar, it defaults to directing you to the Windows Store for the current version of Python. However, the path of the Python installed will not be added to the `PATH` environment variable. It is therefore recommended to download the Python installer from python.org for installation. 

If you have already installed Python through the Windows Store, the installation will run automatically. You can check this installation using `where python` in Command Prompt, and you will see that a `python.exe` is in the `C:\Users\[user]\AppData\Local\Microsoft\WindowsApps\` directory:

```bash
C:\Users\[user]> where python
C:\Users\[user]\AppData\Local\Programs\Python\Python312\python.exe    ### python.org installer installation
C:\Users\[user]\AppData\Local\Microsoft\WindowsApps\python.exe        ### Windows Store installation
```

We can check the newly installed Python version by going into the location and running the Python executable to see the version. 

```bash
cd \Users\[user]\AppData\Local\Microsoft\WindowsApps 
C:\Users\[user]\AppData\Local\Microsoft\WindowsApps>.\python.exe
Python 3.13.7 (tags/v3.13.7:bcee1c3, Aug 14 2025, 14:15:11) [MSC v.1944 64 bit (AMD64)] on win32
Type "help", "copyright", "credits" or "license" for more information.
>>> 
```

Since Python 3.12.2 is an earlier version, installing it by downloading the Python installer from [python.org](https://python.org) is preferred. Simply navigate to the download section of the site, search for the desired version, download the installer, and double-click it to initiate the installation. Make sure that you check the options below when installing:

  - Use admin privileges when installing py.exe (this enables py.exe to manage all Python installations)
  - Add python.exe to the PATH environment variable (this sets this installation as the default Python)

In Windows Command Prompt, you may see:
```bash
C:\Users\[user]>where python
C:\Users\[user]\AppData\Local\Programs\Python\Python313\python.exe
C:\Users\[user]\AppData\Local\Programs\Python\Python312\python.exe
C:\Users\[user]\AppData\Local\Microsoft\WindowsApps\python.exe
```

## Choose Python Version

If you have more than one Python version installed, you may need to specify the version of Python you use for your projects. If the default Python is different from the desired version, you may use the version **suffixes** to designate the desired Python version. This helps, e.g., when you need to create a virtual environment, and you need to designate a Python version such as 3.12.

::::{tab-set}
:::{tab-item} Windows
In Windows, you may use the Python launcher to decide which Python version to use. This means, instead of issuing the `python` command, you use `py -Major.Minor` to control which Python version to use. For example, to use version 3.12, you issue:

```bash
C:\Users\[user]>py -3.12
Python 3.12.2 (tags/v3.12.2:6abddd9, Feb  6 2024, 21:26:36) 
[MSC v.1937 64 bit (AMD64)] on win32
Type "help", "copyright", "credits" or "license" for more information.
>>>
```
:::

:::{tab-item} macOS
In macOS, we append the Python version number in the format of `Major.Minor` when issuing `python` to choose the Python version to use:

```bash
$ python3.12
Python 3.12.2 (v3.12.2:6abddd9f6a, Feb  6 2024, 17:02:06) 
[Clang 13.0.0 (clang-1300.0.29.30)] on darwin
Type "help", "copyright", "credits" or "license" for more information.
>>> 
```
:::
::::

## Set Default Python Version

You can also change the system default Python version for the command `python` by modifying the path entries in the operating system's PATH environment variable (restarting the computer may be needed).

::::{tab-set}
:::{tab-item} Windows

In Windows, you want to move the desired Python entry to the top of the PATH environment variable:

1. Search for "Edit the system environment variables", and you will see the "System Properties" window. 
2. Click on the "Environment Variables" button near the bottom of the window.
4. In the Environment Variables window, click to highlight "Path" in the "User variables for [USER]" pane.
5. Click the "Edit" button to bring up the "Edit environment variables" window.
6. Using the Move Up button on the right, move the Python version entry (e.g., Python312) to the top of the list. 

```{figure} ../../figures/python_EV_order.png
---
width: 450px
name: python_EV-order
---
The top Python path is the default Python
```

This will point the default `python` to your desired version; in this case, Python 3.12:

```bash
PS \\Users\[user]> python
Python 3.12.2 (tags/v3.12.2:6abddd9, Feb  6 2024, 21:26:36) 
[MSC v.1937 64 bit (AMD64)] on win32
Type "help", "copyright", "credits" or "license" for more information.
>>>
```
:::

:::{tab-item} macOS
If you are using macOS, you may change the PATH environment variable by following these steps: 

1. Check Python installation and the default Python version. The example below shows that Linux database does not recognize `python` but `python3` and `python3.12` are recognized by the system. 
```bash
  [user]@mac:~$ which python
  [user]@mac:~$ which python3
  /opt/homebrew/bin/python3
  [user]@mac:~$ which python3.12
  /opt/homebrew/bin/python3.12
  [user]@mac:~$ 
```
2. Edit shell configuration (~/.bashrc or ~/.zshrc)
```bash
  nano ~/.zshrc       ### if using zsh
  nano ~/.bashrc      ### if using bash
```
3. Add preferred Python version to PATH
```bash
  export PATH="/usr/local/bin/python3.12:$PATH"
```

or, if installed via Homebrew:

```bash
  export PATH="/opt/homebrew/bin:$PATH"
```
4. Apply changes
```bash
  source ~/.zshrc       ### if using zsh
  source ~/.bashrc      ### if using bash
```

Or, you may simply set an alias in one of your profile files (such as ~/.bashrc) by adding the line:
```bash
alias python='python3.12'
```
:::
