# Collections

This chapter presents Python collections for storing multiple values: lists, tuples, and sets. You will compare ordered, mutable, immutable, and unique-value containers so you can choose the right structure for a task.

- Creating and using lists, tuples, and sets
- Accessing ordered collections with indexes and slices
- Adding, removing, and transforming mutable collections
- Comparing mutability, aliasing, copying, and uniqueness
- Applying common collection idioms used in everyday Python

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Collection | An object that stores multiple values. |
| 2 | Sequence | An ordered collection whose items can be accessed by position. |
| 3 | List | An ordered, mutable sequence of values. |
| 4 | Tuple | An ordered, immutable sequence of values. |
| 5 | Set | An unordered collection of unique values. |
| 6 | Mutability | Whether an object can be changed after it is created. |
| 7 | Immutable | Describes an object whose contents cannot be changed after creation. |
| 8 | Index | An integer position used to access an item in a sequence. |
| 9 | Slice | A selected range of items from a sequence. |
| 10 | List comprehension | A compact expression for building a list from an iterable. |
| 11 | Nested list | A list that contains other lists as elements. |
| 12 | Aliasing | Two or more names referring to the same object. |
| 13 | Shallow copy | A copy of a collection that still shares nested objects with the original. |
| 14 | Deep copy | A copy that recursively copies nested objects. |
| 15 | Unpacking | Assigning items from a collection to separate names. |
| 16 | `enumerate` | A built-in function that produces index-value pairs during iteration. |
| 17 | `zip` | A built-in function that pairs items from multiple iterables. |
| 18 | Hashable | Describes an object whose hash value does not change during its lifetime. |
| 19 | Frozenset | An immutable version of a set. |
| 20 | Set operation | An operation such as union, intersection, difference, or symmetric difference. |
