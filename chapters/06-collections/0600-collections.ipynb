# Sequences

This chapter presents Python's sequence types for storing values in order: lists and tuples. You will compare mutable and immutable sequences, and see how aliasing and copying affect programs that share lists, so you can choose the right structure for a task.

- Creating and using lists and tuples
- Accessing sequences with indexes and slices
- Adding, removing, and transforming lists
- Comparing mutability, identity, aliasing, and copying
- Applying common sequence idioms used in everyday Python

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Collection | An object that stores multiple values. |
| 2 | Sequence | An ordered collection whose items can be accessed by position. |
| 3 | List | An ordered, mutable sequence of values. |
| 4 | Tuple | An ordered, immutable sequence of values. |
| 5 | Mutability | Whether an object can be changed after it is created. |
| 6 | Immutable | Describes an object whose contents cannot be changed after creation. |
| 7 | Index | An integer position used to access an item in a sequence. |
| 8 | Slice | A selected range of items from a sequence. |
| 9 | List comprehension | A compact expression for building a list from an iterable. |
| 10 | Nested list | A list that contains other lists as elements. |
| 11 | Aliasing | Two or more names referring to the same object. |
| 12 | Shallow copy | A copy of a collection that still shares nested objects with the original. |
| 13 | Deep copy | A copy that recursively copies nested objects. |
| 14 | Unpacking | Assigning items from a collection to separate names. |
| 15 | `enumerate` | A built-in function that produces index-value pairs during iteration. |
| 16 | `zip` | A built-in function that pairs items from multiple iterables. |
| 17 | Identity | The property that distinguishes one object from another; `is` checks whether two names refer to the same object. |
| 18 | Reference | The association between a variable and the object it refers to. |
