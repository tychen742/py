Some values belong together and should not change: a location's latitude and longitude, a date's year, month, and day, or one row of a sales report. A **tuple** stores values like these as a single, fixed sequence. Section 6.2 showed that a list can change through any name that refers to it; a tuple cannot change at all, which makes it safe to share and a natural way to represent records.

This section covers what tuples are, how to create and access them, what immutability does and does not guarantee, how to unpack tuples, the operations tuples support, how functions use tuples, and how to work with tuples as records.

## What Tuples Are

There are two common pronunciations of "tuple". Some people say "tuh-ple" (rhymes with "supple"), and many programmers say "too-ple" (rhymes with "quadruple").

- A tuple is a **sequence** of values.
- The values can be **any type**.
- The values in a tuple are **indexed** by integers, so tuples feel a lot like lists.
- When a function returns multiple values, it returns a tuple.
- Because tuples can't change, they can be used in places that require fixed values, such as dictionary keys (Chapter 7).

For example, a latitude and longitude pair is a natural tuple:

````python
location = (42.3601, -71.0589)
````

Use a tuple when the values belong together and should not change; use a list when you need to add, remove, or change items. The table compares the two:

| Feature    | Tuple          | List           |
|------------|----------------|----------------|
| Syntax     | `(1, 2, 3)`    | `[1, 2, 3]`    |
| Mutable    | No             | Yes            |
| Methods    | 2 (`.count`, `.index`) | Many  |
| Use case   | Fixed data, records | Collections that change |

## Creating and Accessing Tuples

Tuples in Python can be created in several ways, primarily: 
- Using tuple literals: through the use of commas and optional parentheses
- by using the built-in tuple() constructor

| How to create | Method | Example | Result |
|---|---|---|---|
| Use parentheses with comma-separated values | Parentheses | `(1, 2, 3)` | `(1, 2, 3)` |
| Write comma-separated values without parentheses | Tuple packing | `1, 2, 3` | `(1, 2, 3)` |
| Use empty parentheses | Empty tuple | `()` | `()` |
| Add a trailing comma for one value | Single-item tuple | `(1,)` or `1,` | `(1,)` |
| Use the tuple constructor on an iterable | Constructor | `tuple([1, 2, 3])` | `(1, 2, 3)` |

### Tuple Literals

To create a tuple, you can write a comma-separated list of values.

````python
status = 'NEW', 'PROCESSING', 'SHIPPED', 'DELIVERED', 'CLOSED'
type(status)
````

Parentheses are optional, but it is common to use them with tuples.

````python
t1 = ('NEW', 'PROCESSING', 'SHIPPED', 'DELIVERED', 'CLOSED')
print(t1)
print(type(t1))

t2 = 'NEW', 'PROCESSING', 'SHIPPED', 'DELIVERED', 'CLOSED'  # tuple packing
print(t2)
print(type(t2))
````

### Single-Element Tuples

To make a one-element tuple, include a trailing **comma**. Parentheses are optional.

````python
t1 = ('URGENT',)
print(type(t1))
print(t1[0])

t2 = 'URGENT',               # tuple packing with one element
print(type(t2))
print(t2[0])
````

However, a single value inside parentheses is not a tuple without a comma.

````python
t2 = ('URGENT')
type(t2)
````

### `tuple()` Constructor

Another way to create a tuple is with the built-in function/type constructor `tuple`.

If the argument is any iterable, such as a list, a string, or a range, `tuple` returns a tuple containing its elements. The `list` constructor works the same way in reverse, turning a tuple into a list.

````python
t = tuple(['Laptop', 'Mouse', 'Keyboard', 'Monitor'])
print(t)
print(type(t))

l = list(('Laptop', 'Mouse', 'Keyboard', 'Monitor'))
print(l)
print(type(l))
````

With no argument, `tuple()` returns an empty tuple. Empty parentheses do the same:

````python
t = tuple()
print(t, type(t))

empty = ()
print(empty, type(empty))
````

### Nested Tuples

A **nested tuple** is a tuple that contains one or more tuples as elements. This is useful for representing multi-dimensional or grouped data, such as a matrix or a list of coordinate pairs. You access elements using chained indexing: the first index selects the inner tuple, and the second index selects an element within it.

````python
# A tuple containing other tuples
matrix = ((1, 2, 3), (4, 5, 6), (7, 8, 9))

# Access the second row
print(matrix[1])        # (4, 5, 6)

# Access the third element of the second row
print(matrix[1][2])     # 6

# Tuple of coordinate pairs
points = ((0, 0), (1, 2), (3, 4))
for x, y in points:
    print(f'x={x}, y={y}')
````

## Immutability

Tuples are **immutable**: once a tuple is created, its items cannot be added, removed, or replaced. Operations such as `+` build a new tuple instead of changing the old one, which is why tuples have no methods like `append` or `remove`.

### Tuples Cannot Change

If you try to replace an item in a tuple, Python raises a `TypeError`.

````python
%%expect TypeError
invoice_key = ("INV-2026-0142", "ACME", "2026-02-28")
print("Invoice key[0]:", invoice_key[0])

invoice_key[0] = "INV-2026-0143" # Tuple does not support item assignment
````

To get a tuple with a different value, build a new one. Slicing and `+` work on tuples the same way they work on lists:

````python
point = (3, 4, 5)
new_point = point[:2] + (0,)    # a new tuple; point is unchanged
print(point, new_point)
````

### Tuples That Contain Lists

Immutability applies to the tuple itself: which objects it holds, and in what order. If one of those objects is mutable, such as a list, that object can still change. This is the same idea as the shallow copies in Section 6.2: the outer container is fixed, but the list inside it is a shared object that can be modified.

````python
record = ('Ana', [91, 84])     # a name and a list of scores
scores = record[1]
scores.append(77)              # changes the list inside the tuple

print(record)
print(scores is record[1])     # True: the same list object
````

The tuple still holds the same two objects, but the list it holds has changed. Replacing the list itself is still an error:

````python
%%expect TypeError
record[1] = [100]
````

If a record must never change, store only immutable values inside it, such as numbers, strings, and other tuples: `('Ana', (91, 84))`.

### Safe to Share

Section 6.2 showed that passing a list to a function passes a reference, so the function can change the caller's list. A tuple cannot be changed, so passing one to a function is safe: the function can read it but not modify it.

````python
def add_bonus(scores):
    scores[0] = scores[0] + 5      # works for a list, fails for a tuple
    return scores

list_scores = [80, 90]
add_bonus(list_scores)
print(list_scores)                 # the caller's list changed
````

````python
%%expect TypeError
tuple_scores = (80, 90)
add_bonus(tuple_scores)
````

To produce updated values from a tuple, a function returns a new tuple instead of changing the old one:

````python
def with_bonus(scores):
    return tuple([score + 5 for score in scores])

tuple_scores = (80, 90)
new_scores = with_bonus(tuple_scores)
print(tuple_scores, new_scores)
````

This is a common design choice: when the code that receives some data should not change it, store the data in a tuple.

## Unpacking and Iteration Helpers

### Basic Unpacking

You can assign multiple variables from a tuple or any sequence.

````python
a, b = 1, 2
````

Values are assigned left to right. In this example, `a` gets `1` and `b` gets `2`.
You can display the results like this:

````python
a, b
````

More generally, if the left side is a tuple of variables, the right side can be any sequence - a string, list, or tuple.
For example, to split an email address into a user name and domain:

````python
email = 'monty@python.org'
username, domain = email.split('@')
````

`split` returns a list with two elements; the first goes to `username`, the second to `domain`.

````python
username, domain
````

The number of variables on the left and values on the right must **match**; otherwise you get a `ValueError`.

````python
%%expect ValueError
a, b = 1, 2, 3
````

Tuple assignment is a clean way to swap two variables.
With conventional assignments, you need a temporary variable:

````python
temp = a
a = b
b = temp
````

That works, but tuple assignment does the same swap without a temporary variable. It works because Python evaluates the whole right side, `b, a`, before it assigns anything.

````python
a, b = b, a
````

### `zip()`

Tuples are useful for pairing elements from multiple sequences and working with them together.
For example, suppose a sales team has weekly targets, and their actual sales and the targets are stored in two separate lists.

````python
actual = [112, 98, 135, 144, 101, 129, 88]
target = [120, 120, 120, 120, 120, 120, 120]
````

To count how many weeks the team exceeded its target, use `zip`, which you saw in Section 6.1. It combines sequences and returns a **zip object** that pairs elements like the teeth of a zipper, and each pair is a tuple.

````python
type(zip(actual, target))
````

`zip` stops when the shortest input runs out. To keep going, use `itertools.zip_longest`; its `fillvalue` argument sets the value used for the missing items.

````python
from itertools import zip_longest

list(zip('abc', [1, 2]))
list(zip_longest('abc', [1, 2], fillvalue='-'))
````

You can loop over the zip object to get the pairs.

````python
for pair in zip(actual, target):
    print(pair)
````

Each time through the loop, `pair` is a tuple of `(actual_sales, target_sales)`.
You can unpack those values and count the weeks where actual sales exceeded the target:

````python
weeks_above_target = 0
for sale, goal in zip(actual, target):
    if sale > goal:
        weeks_above_target += 1

weeks_above_target
````

The team beat their weekly target three out of seven weeks.

If you want a list of pairs, combine `zip` with `list`.

````python
weekly_results = list(zip(actual, target))
weekly_results
````

The result is a list of tuples, so you can get the last week's data like this:

````python
weekly_results[-1]
````

## Tuple Operations

Common tuple operations include: 
| Python Expression | Description | Example |
|---|---|---|
| `len(tuple)` | Returns the number of items in the tuple. | `len((1, 2, 3))` returns `3`. |
| `tuple1 + tuple2` | Concatenation: combines two tuples into a new tuple. | `(1, 2) + (3, 4)` returns `(1, 2, 3, 4)`. |
| `tuple * n` | Repetition: repeats tuple elements `n` times. | `('Hi!',) * 3` returns `('Hi!', 'Hi!', 'Hi!')`. |
| `item in tuple` | Membership: checks whether a value exists in the tuple (`True`/`False`). | `3 in (1, 2, 3)` returns `True`. |
| `tuple[index]` | Indexing: accesses an element using positive or negative index. | `('a', 'b', 'c')[0]` returns `'a'`. |
| `tuple[start:stop]` | Slicing: extracts a range of elements as a new tuple. | `(1, 2, 3, 4)[1:3]` returns `(2, 3)`. |
| `tuple(iterable)` | Constructor: converts an iterable (list, string, range, etc.) into a tuple. | `tuple([1, 2])` returns `(1, 2)`. |

### Tuple Functions

In addition to these two methods, Python's built-in functions (e.g., `len`, `max`, `min`, `sum`) can be used with tuples, as they work with any `iterable`. These functions do not **modify** the original tuple, but return **new** values or objects: 

| Function | What it does | Notes |
|---|---|---|
| `len(tuple)` | Returns the total number of items in the tuple. | Works for any tuple. |
| `max(tuple)` | Returns the largest item in the tuple. | Elements must be comparable. |
| `min(tuple)` | Returns the smallest item in the tuple. | Elements must be comparable. |
| `sum(tuple)` | Returns the sum of all numeric elements in the tuple. | Elements should be numbers. |
| `sorted(tuple)` | Returns a new list with tuple elements in sorted order. | Output type is **`list`**, not `tuple`. |
| `reversed(tuple)` | Returns an iterator over tuple elements in reverse order. | Convert with `tuple(...)` or `list(...)` if needed. |
| `tuple(iterable)` | Converts another iterable (like a list or string) into a tuple. | Useful for type conversion. |

The `sorted` function works with tuples, but it returns a new `list`, not a tuple. The original tuple is unchanged.

### Comparing and Sorting

Comparison operators such as `<` and `==` work with tuples and other sequences.
Tuple comparison is lexicographic: it compares the first elements, then the next, and so on until it finds a difference.
For example, you can represent a reporting period as a `(year, quarter)` tuple and compare two periods.

````python
(2026, 1) < (2026, 3)    # Q1 2026 is before Q3 2026
````

Once a difference is found, later elements are not considered.

````python
(2026, 1, 999_999) < (2026, 3, 62_000)   # Q1 beats Q3 in revenue, but Q3 comes later
````

This comparison behavior is useful for sorting lists of tuples or finding minimum and maximum values.

````python
### Exercise: Comparing and Sorting *(basic)*
#   1. Given a list of (last_name, first_name) tuples below,
#      sort it lexicographically (default tuple sort) and print the result.
#   2. What does Python compare first: last name or first name?
people = [('Smith', 'John'), ('Adams', 'Zara'), ('Smith', 'Alice'), ('Adams', 'Alan')]
### Your code starts here.




### Your code ends here.
````

### Sorting with a Key Function

Sorting a list of tuples compares the first elements first. To sort by a different field, pass a function as the optional `key` argument to `sorted`. The `key` function computes a **sort key** for each item.

````python
records = [('Dee', 95), ('Bob', 84), ('Ana', 91), ('Chen', 77)]
sorted(records)          # sorts by name, the first element
````

(second_element)=

````python
def second_element(t):
    return t[1]

sorted(records, key=second_element)    # sorts by score
````

The lowest score comes first and the highest comes last. If you only want the maximum, you don't need to sort: `max` also accepts a `key` function.

````python
max(records, key=second_element)
````

## Tuples in Function Design

### Tuples as Return Values

**A function returns one object**, but that object can be a tuple.
Returning a tuple is a common way to provide **multiple** results.

For example, a scheduling system may need to convert a total number of minutes into hours and minutes.
The built-in function `divmod` combines `//` and `%`: it takes two numbers and returns a tuple with the quotient and the remainder.

````python
divmod(135, 60)    # 135 minutes → (2 hours, 15 minutes)
````

You can use tuple assignment to store the hours and minutes in separate variables.

````python
hours, minutes = divmod(135, 60)
hours
````

````python
minutes
````

Here is a simple function that returns both the minimum and maximum from a list, which is useful for summarizing a range of values such as daily sales figures.

````python
def min_max(t):
    return min(t), max(t)
````

`min` and `max` are built-in functions that find the smallest and largest elements of a sequence.
`min_max` computes both and returns them as a tuple.

### Argument Packing

Functions can take a variable number of arguments.
A parameter name that begins with `*` **packs** extra arguments into a tuple.
The following function takes any number of arguments and computes their arithmetic mean. (Python's `statistics` module already has a `mean` function; writing one here shows how packing works.)

````python
def mean(*args):                # variable-length argument list
    return sum(args) / len(args)
````

The parameter name can be anything, but `args` is conventional.
Here is a call with four daily sales figures:

````python
mean(8500, 12400, 7200, 9900)    # average of four daily sales figures
````

If you have a sequence of values and want to pass them as separate arguments, use `*` to **unpack** the sequence.
For example, `divmod` takes exactly two arguments - if you pass a tuple, it counts as a single argument and raises an error.

````python
%%expect TypeError
t = (135, 60)
divmod(t)
````

Even though the tuple contains two elements, it is still one argument.
If you unpack it, the two elements are passed separately.

````python
divmod(*t)
````

Packing and unpacking are handy when you want to adapt an existing function.
For example, this function takes any number of arguments, removes the lowest and highest, and computes the mean of the rest.

````python
def trimmed_mean(*args):
    low, high = min_max(args)
    trimmed = list(args)        # convert to a list for mutability
    trimmed.remove(low)
    trimmed.remove(high)
    return mean(*trimmed)
````

First it uses `min_max` to find the lowest and highest values.
Then it converts `args` to a list so it can use `remove`.
Finally it **unpacks** the trimmed list so the elements are passed to `mean` as separate arguments.

Here is an example using judge scores, where one score is unusually low:

````python
mean(8.5, 9.2, 8.8, 4.5)    # raw mean including the outlier
````

````python
trimmed_mean(8.5, 9.2, 8.8, 4.5)    # mean after dropping lowest (4.5) and highest (9.2)
````

This kind of trimmed mean is used in sports with subjective judging (like diving and gymnastics) to reduce the influence of outlier scores.

You can also unpack sequences to build new tuples (or lists) without using `+`.
For example, merging two quarterly sales tuples into a single half-year tuple:

````python
q1 = (112, 98, 135)    # monthly sales for Q1
q2 = (144, 101, 129)   # monthly sales for Q2
h1 = (*q1, *q2)        # combined first-half data
h1
````

## Tuples as Records

A **record** is a group of related values that describe one thing, such as one sale or one student. Tuples fit records well: the fields stay together, their order is fixed, and the record cannot be changed by accident. A list of tuples works like a small table, with one tuple per row.

### Lists of Tuples

Data often arrives as lines of text. Each line below describes one sale: the region, the month, and the amount.

````python
lines = [
    'North,Jan,112',
    'South,Jan,98',
    'North,Feb,135',
    'West,Jan,144',
    'South,Feb,101',
]

sales = []
for line in lines:
    region, month, amount = line.split(',')
    sales.append((region, month, int(amount)))

sales
````

Each row is now a `(region, month, amount)` tuple. You can loop over the rows with unpacking:

````python
for region, month, amount in sales:
    print(region, month, amount)
````

You can also filter the rows with a list comprehension and sort them with a `key` function:

````python
north = [row for row in sales if row[0] == 'North']
print(north)

def amount_of(row):
    return row[2]

print(sorted(sales, key=amount_of)[-2:])    # the two largest sales, smallest first
````

````python
### Exercise: Lists of Tuples *(basic)*
#   Given lines = ['Ana,91', 'Bob,84', 'Chen,77', 'Dee,95']:
#   1. Build a list of (name, score) tuples called students,
#      converting each score to an int.
#   2. Use a list comprehension to print the names of students
#      who scored 85 or above.
#   3. Print the student with the lowest score using min() with a key function.
lines = ['Ana,91', 'Bob,84', 'Chen,77', 'Dee,95']
### Your code starts here.



### Your code ends here.
````

````python
### Solution
lines = ['Ana,91', 'Bob,84', 'Chen,77', 'Dee,95']

students = []
for line in lines:
    name, score = line.split(',')
    students.append((name, int(score)))

print([name for name, score in students if score >= 85])

def score_of(student):
    return student[1]

print(min(students, key=score_of))
````

### Named Tuples

Positions such as `row[2]` are easy to mix up: nothing in the code says that index 2 is the amount. A **named tuple** is a tuple whose fields also have names. The `namedtuple` function from the `collections` module creates a new tuple type with the field names you choose.

````python
from collections import namedtuple

Sale = namedtuple('Sale', ['region', 'month', 'amount'])

s = Sale('North', 'Jan', 112)
print(s)
print(s.amount)       # access by name
print(s[2])           # access by position still works
````

A named tuple is still a tuple: it supports indexing, unpacking, and comparison. Converting the rows makes the code easier to read:

````python
sale_rows = [Sale(region, month, amount) for region, month, amount in sales]

for s in sale_rows:
    if s.amount > 100:
        print(s.region, s.month, s.amount)
````

Like any tuple, a named tuple cannot be changed:

````python
%%expect AttributeError
s.amount = 200
````

You will meet named tuples again in data work: the pandas method `DataFrame.itertuples()` returns each row of a table as a named tuple.

````python
### Exercise: Named Tuples *(applied)*
#   1. Create a named tuple type Student with the fields name and score.
#   2. Convert students = [('Ana', 91), ('Bob', 84), ('Chen', 77), ('Dee', 95)]
#      into a list of Student named tuples.
#   3. Print the name of the top student using max() with a key function
#      that returns the score field.
students = [('Ana', 91), ('Bob', 84), ('Chen', 77), ('Dee', 95)]
### Your code starts here.



### Your code ends here.
````

````python
### Solution
from collections import namedtuple

Student = namedtuple('Student', ['name', 'score'])
students = [('Ana', 91), ('Bob', 84), ('Chen', 77), ('Dee', 95)]

roster = [Student(name, score) for name, score in students]

def score_of(student):
    return student.score

print(max(roster, key=score_of).name)
````

### Worked Example: Summarizing Sales Records

The task is to report the total and average sale, the smallest and largest sale, and the two largest sales from `sale_rows`, using a function that returns several results as a tuple.

````python
def amount_key(sale):
    return sale.amount

def summarize(rows):
    amounts = [sale.amount for sale in rows]
    total = sum(amounts)
    average = total / len(amounts)
    smallest = min(rows, key=amount_key)
    largest = max(rows, key=amount_key)
    return total, average, smallest, largest

total, average, smallest, largest = summarize(sale_rows)
top_two = sorted(sale_rows, key=amount_key)[::-1][:2]

print(f'Total: {total}')
print(f'Average: {round(average, 1)}')
print(f'Smallest: {smallest.region} {smallest.month} {smallest.amount}')
print(f'Largest: {largest.region} {largest.month} {largest.amount}')
print('Top two:', [(s.region, s.month) for s in top_two])
````

Each step uses a tool from this section:

- Named tuples make fields such as `sale.amount` readable.
- A list comprehension collects the amounts, and `sum()` and `len()` compute the total and average.
- `min()` and `max()` with a `key` function find whole records, not just amounts.
- `summarize` returns four results as one tuple, and tuple assignment unpacks them.
- `sorted()` with a `key`, followed by slicing, ranks the records.

## Summary

- A tuple is an ordered, immutable sequence. Create one with commas (a trailing comma for a single item) or with `tuple()`.
- Indexing, slicing, `+`, `*`, `in`, `count()`, `index()`, and built-in functions such as `len()` and `sorted()` work with tuples.
- A tuple's items cannot be replaced, but a mutable item inside it, such as a list, can still change.
- Tuples are safe to pass to functions; to produce new values, return a new tuple.
- Tuple assignment and starred unpacking assign several names at once, and `zip()` and `enumerate()` produce tuples.
- Tuples compare element by element; `sorted()`, `min()`, and `max()` accept a `key` function to compare by a chosen field.
- A function can return several results as a tuple, and `*args` packs any number of arguments into a tuple.
- A list of tuples works like a table of records, and named tuples give each field a name.
