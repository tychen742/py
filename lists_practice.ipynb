## 3. Updateing Lists

### 3.1. List Operators

```python
nums1 = [ num for num in range(10) if num%2 == 0]
nums2 = [ num for num in range(10) if num%2 != 0]

print(nums1)
print(nums2)
```

```python
nums1 + nums2
```

```python
nums1 * 3
```

```python
nums1 - nums2
```
